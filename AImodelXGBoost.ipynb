```python
%%writefile scripts/91_modeling/train_baseline_v1_final_ensemble.py

from pathlib import Path
import json

import pandas as pd
import xgboost as xgb


ROOT = Path(__file__).resolve().parents[2]

MATRIX_DIR = ROOT / "output" / "baseline_v1_model_matrices"

OUT_DIR = ROOT / "output" / "xgboost_baseline_v1_final_ensemble"
OUT_DIR.mkdir(parents=True, exist_ok=True)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",
    "Distance_to_Road_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


PARAMS = {
    "objective": "binary:logistic",
    "n_estimators": 300,
    "learning_rate": 0.05,
    "max_depth": 4,
    "min_child_weight": 1.0,
    "gamma": 0.0,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "reg_alpha": 0.0,
    "reg_lambda": 1.0,
    "scale_pos_weight": 1.0,
    "tree_method": "hist",
    "device": "cpu",
    "eval_metric": "logloss",
    "random_state": 20260816,
    "n_jobs": -1,
    "verbosity": 0,
    "validate_parameters": True,
}


print("=" * 90)
print("MANIPUR XGBOOST BASELINE V1")
print("FINAL FIVE-MODEL ENSEMBLE TRAINING")
print("=" * 90)

print("Feature count :", len(FEATURES))
print("Training sets : 5")
print("Parameters    : Frozen Baseline-v1")
print()


manifest = {
    "model_name": "Manipur XGBoost Baseline-v1 Final Ensemble",
    "number_of_models": 5,
    "feature_count": len(FEATURES),
    "features": FEATURES,
    "parameters": PARAMS,
    "models": [],
}


for set_id in range(1, 6):

    matrix_path = MATRIX_DIR / (
        f"manipur_baseline_v1_model_matrix_set_{set_id:02d}.csv"
    )

    print()
    print("-" * 90)
    print(f"FINAL MODEL {set_id}/5")
    print("-" * 90)
    print("Input:", matrix_path)

    if not matrix_path.exists():
        raise FileNotFoundError(
            f"Training matrix not found: {matrix_path}"
        )

    df = pd.read_csv(matrix_path)

    missing_features = [
        feature for feature in FEATURES
        if feature not in df.columns
    ]

    if missing_features:
        raise RuntimeError(
            f"Missing features in Set {set_id:02d}: "
            f"{missing_features}"
        )

    if "Label" not in df.columns:
        raise RuntimeError(
            f"Label column missing in Set {set_id:02d}"
        )

    X = df[FEATURES].copy()
    y = df["Label"].astype(int)

    if X.isna().any().any():
        raise RuntimeError(
            f"Missing predictor values detected "
            f"in Set {set_id:02d}"
        )

    n_rows = len(df)
    n_positive = int((y == 1).sum())
    n_background = int((y == 0).sum())

    print("Rows       :", n_rows)
    print("Positive   :", n_positive)
    print("Background :", n_background)
    print("Features   :", X.shape[1])

    if n_rows != 4544:
        raise RuntimeError(
            f"Unexpected number of rows: {n_rows}"
        )

    if n_positive != 1136:
        raise RuntimeError(
            f"Unexpected positive count: {n_positive}"
        )

    if n_background != 3408:
        raise RuntimeError(
            f"Unexpected background count: {n_background}"
        )

    if X.shape[1] != 25:
        raise RuntimeError(
            f"Unexpected feature count: {X.shape[1]}"
        )

    model = xgb.XGBClassifier(**PARAMS)

    model.fit(X, y)

    model_path = OUT_DIR / (
        f"manipur_xgb_baseline_v1_final_set_{set_id:02d}.json"
    )

    model.save_model(model_path)

    print("Model saved:", model_path)

    manifest["models"].append({
        "dataset_set": set_id,
        "training_rows": n_rows,
        "positive_rows": n_positive,
        "background_rows": n_background,
        "model_file": model_path.name,
    })


manifest_path = OUT_DIR / "final_ensemble_manifest.json"

with open(manifest_path, "w") as f:
    json.dump(manifest, f, indent=2)


feature_path = OUT_DIR / "final_feature_order.json"

with open(feature_path, "w") as f:
    json.dump(FEATURES, f, indent=2)


print()
print("=" * 90)
print("FINAL ENSEMBLE TRAINING COMPLETE ✅")
print("=" * 90)

print("Number of models:", len(manifest["models"]))
print("Feature count   :", len(FEATURES))
print("Output folder   :", OUT_DIR)
```

```python
%%writefile /content/Manipur_XGBoost_Colab/scripts/91_modeling/predict_baseline_v1_statewide_ensemble.py

from pathlib import Path
from contextlib import ExitStack
import json
import math
import time

import numpy as np
import rasterio
from rasterio.windows import Window
import xgboost as xgb


# =============================================================================
# PATHS
# =============================================================================

PROJECT_DIR = Path("/content/Manipur_XGBoost_Colab")

RASTER_DIR = Path("/content/Manipur_XGBoost_Statewide_Rasters")

MODEL_DIR = (
    PROJECT_DIR
    / "output"
    / "xgboost_baseline_v1_final_ensemble"
)

OUT_DIR = (
    PROJECT_DIR
    / "output"
    / "statewide_prediction_baseline_v1"
)

OUT_DIR.mkdir(parents=True, exist_ok=True)


# =============================================================================
# FROZEN MODEL FEATURE ORDER
# =============================================================================

FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",
    "Distance_to_Road_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


LITHOLOGY_CODES = list(range(1, 12))

LULC_CODES = [
    10,
    20,
    30,
    40,
    50,
    60,
    80,
    90,
]


# =============================================================================
# INPUT RASTERS
# =============================================================================

RASTER_FILES = {
    "Elevation_m":
        "manipur_dem_30m_utm46n.tif",

    "Slope_deg":
        "manipur_slope_30m.tif",

    "TWI":
        "manipur_twi_30m.tif",

    "Rainfall_CHELSA_Annual_mm":
        "manipur_rainfall_annual_30m.tif",

    "Distance_to_River_m":
        "manipur_distance_to_river_30m.tif",

    "Distance_to_Road_m":
        "manipur_distance_to_road_30m.tif",

    "Lithology_Group_Code":
        "manipur_lithology_grouped_30m.tif",

    "LULC_WorldCover_Code":
        "manipur_lulc_worldcover_2021_30m.tif",
}


MASK_PATH = (
    RASTER_DIR
    / "manipur_baseline_common_valid_mask_30m.tif"
)


# =============================================================================
# OUTPUTS
# =============================================================================

MEAN_PATH = (
    OUT_DIR
    / "manipur_xgb_baseline_v1_susceptibility_mean.tif"
)

SD_PATH = (
    OUT_DIR
    / "manipur_xgb_baseline_v1_prediction_sd.tif"
)

MANIFEST_PATH = (
    OUT_DIR
    / "statewide_prediction_manifest.json"
)


WINDOW_SIZE = 512

OUTPUT_NODATA = -9999.0

EXPECTED_VALID_CELLS = 24_718_359


# =============================================================================
# HELPERS
# =============================================================================

def make_windows(height, width, size):
    for row_off in range(0, height, size):

        h = min(size, height - row_off)

        for col_off in range(0, width, size):

            w = min(size, width - col_off)

            yield Window(
                col_off=col_off,
                row_off=row_off,
                width=w,
                height=h,
            )


def same_grid(a, b):
    return (
        a.crs == b.crs
        and a.shape == b.shape
        and a.res == b.res
        and a.transform == b.transform
        and a.bounds == b.bounds
    )


# =============================================================================
# HEADER
# =============================================================================

print("=" * 100)
print("MANIPUR XGBOOST BASELINE V1")
print("STATEWIDE FIVE-MODEL ENSEMBLE PREDICTION")
print("=" * 100)

print("Raster directory :", RASTER_DIR)
print("Model directory  :", MODEL_DIR)
print("Output directory :", OUT_DIR)
print("Feature count    :", len(FEATURES))
print("Ensemble models  : 5)
print("Window size      :", WINDOW_SIZE)
print()


# =============================================================================
# VERIFY INPUT FILES
# =============================================================================

if not MASK_PATH.exists():
    raise FileNotFoundError(
        f"Common valid mask not found: {MASK_PATH}"
    )


raster_paths = {}

for name, filename in RASTER_FILES.items():

    path = RASTER_DIR / filename

    if not path.exists():
        raise FileNotFoundError(
            f"Missing raster: {path}"
        )

    raster_paths[name] = path


model_paths = [
    MODEL_DIR
    / f"manipur_xgb_baseline_v1_final_set_{i:02d}.json"
    for i in range(1, 6)
]


for path in model_paths:

    if not path.exists():
        raise FileNotFoundError(
            f"Missing model: {path}"
        )


print("Input file validation: PASS ✅")


# =============================================================================
# LOAD FIVE FINAL MODELS
# =============================================================================

boosters = []

for i, model_path in enumerate(model_paths, start=1):

    booster = xgb.Booster()

    booster.load_model(str(model_path))

    model_features = booster.feature_names

    if (
        model_features is not None
        and model_features != FEATURES
    ):
        raise RuntimeError(
            f"Feature mismatch in model {i}\n"
            f"Expected: {FEATURES}\n"
            f"Model:    {model_features}"
        )

    boosters.append(booster)

    print(
        f"Model {i}/5 loaded ✅ : "
        f"{model_path.name}"
    )


print()
print("All 5 models loaded: PASS ✅")


# =============================================================================
# PREFLIGHT GRID + CATEGORY VALIDATION
# =============================================================================

print()
print("=" * 100)
print("PREFLIGHT VALIDATION")
print("=" * 100)


valid_cell_count = 0

unexpected_lithology = set()

unexpected_lulc = set()


with ExitStack() as stack:

    mask_src = stack.enter_context(
        rasterio.open(MASK_PATH)
    )

    sources = {
        name: stack.enter_context(
            rasterio.open(path)
        )
        for name, path in raster_paths.items()
    }

    print("Reference CRS       :", mask_src.crs)
    print("Reference shape     :", mask_src.shape)
    print("Reference resolution:", mask_src.res)
    print("Reference transform :", mask_src.transform)

    for name, src in sources.items():

        if not same_grid(mask_src, src):

            raise RuntimeError(
                f"Grid mismatch detected: {name}"
            )

        print(f"{name:28s}: GRID PASS ✅")


    windows = list(
        make_windows(
            mask_src.height,
            mask_src.width,
            WINDOW_SIZE,
        )
    )

    print()
    print("Number of raster windows:", len(windows))


    for window in windows:

        mask = mask_src.read(
            1,
            window=window,
        )

        valid = mask > 0

        n_valid = int(valid.sum())

        if n_valid == 0:
            continue

        valid_cell_count += n_valid


        lith = sources[
            "Lithology_Group_Code"
        ].read(
            1,
            window=window,
        )[valid]

        lulc = sources[
            "LULC_WorldCover_Code"
        ].read(
            1,
            window=window,
        )[valid]


        lith_codes = set(
            np.unique(
                np.rint(lith).astype(np.int16)
            ).tolist()
        )

        lulc_codes = set(
            np.unique(
                np.rint(lulc).astype(np.int16)
            ).tolist()
        )


        unexpected_lithology.update(
            lith_codes
            - set(LITHOLOGY_CODES)
        )

        unexpected_lulc.update(
            lulc_codes
            - set(LULC_CODES)
        )


print()
print("Valid cells found:", f"{valid_cell_count:,}")
print(
    "Expected valid   :",
    f"{EXPECTED_VALID_CELLS:,}"
)


if unexpected_lithology:

    raise RuntimeError(
        "Unexpected lithology codes detected: "
        f"{sorted(unexpected_lithology)}"
    )


if unexpected_lulc:

    raise RuntimeError(
        "Unexpected LULC codes detected: "
        f"{sorted(unexpected_lulc)}"
    )


if valid_cell_count != EXPECTED_VALID_CELLS:

    print(
        "WARNING: valid-cell count differs "
        "from frozen expected value."
    )

else:

    print("Valid-cell count: PASS ✅")


print("Lithology categories: PASS ✅")
print("LULC categories      : PASS ✅")
print()
print("PREFLIGHT VALIDATION COMPLETE ✅")


# =============================================================================
# STATEWIDE PREDICTION
# =============================================================================

print()
print("=" * 100)
print("STARTING STATEWIDE PREDICTION")
print("=" * 100)

start_time = time.time()


with ExitStack() as stack:

    mask_src = stack.enter_context(
        rasterio.open(MASK_PATH)
    )

    sources = {
        name: stack.enter_context(
            rasterio.open(path)
        )
        for name, path in raster_paths.items()
    }


    profile = mask_src.profile.copy()

    profile.update(
        driver="GTiff",
        dtype="float32",
        count=1,
        nodata=OUTPUT_NODATA,
        compress="deflate",
        predictor=3,
        tiled=True,
        blockxsize=256,
        blockysize=256,
        BIGTIFF="IF_SAFER",
    )


    mean_dst = stack.enter_context(
        rasterio.open(
            MEAN_PATH,
            "w",
            **profile,
        )
    )

    sd_dst = stack.enter_context(
        rasterio.open(
            SD_PATH,
            "w",
            **profile,
        )
    )


    windows = list(
        make_windows(
            mask_src.height,
            mask_src.width,
            WINDOW_SIZE,
        )
    )

    total_windows = len(windows)

    processed_valid_cells = 0

    global_mean_min = float("inf")
    global_mean_max = float("-inf")

    global_sd_min = float("inf")
    global_sd_max = float("-inf")

    susceptibility_sum = 0.0
    sd_sum = 0.0


    for window_index, window in enumerate(
        windows,
        start=1,
    ):

        mask = mask_src.read(
            1,
            window=window,
        )

        valid = mask > 0


        tile_height = int(window.height)
        tile_width = int(window.width)


        mean_tile = np.full(
            (tile_height, tile_width),
            OUTPUT_NODATA,
            dtype=np.float32,
        )

        sd_tile = np.full(
            (tile_height, tile_width),
            OUTPUT_NODATA,
            dtype=np.float32,
        )


        n_valid = int(valid.sum())


        if n_valid > 0:

            X = np.empty(
                (n_valid, 25),
                dtype=np.float32,
            )


            # ---------------------------------------------------------
            # SIX CONTINUOUS FEATURES
            # ---------------------------------------------------------

            continuous = [
                "Elevation_m",
                "Slope_deg",
                "TWI",
                "Rainfall_CHELSA_Annual_mm",
                "Distance_to_River_m",
                "Distance_to_Road_m",
            ]


            for column_index, feature in enumerate(
                continuous
            ):

                arr = sources[feature].read(
                    1,
                    window=window,
                )

                values = arr[valid].astype(
                    np.float32,
                    copy=False,
                )

                nodata = sources[feature].nodata

                if np.any(~np.isfinite(values)):

                    raise RuntimeError(
                        f"Non-finite values detected "
                        f"in {feature}"
                    )

                if (
                    nodata is not None
                    and np.any(values == nodata)
                ):

                    raise RuntimeError(
                        f"NoData detected inside common "
                        f"valid mask for {feature}"
                    )

                X[:, column_index] = values


            # ---------------------------------------------------------
            # LITHOLOGY → 11 ONE-HOT FEATURES
            # ---------------------------------------------------------

            lith = sources[
                "Lithology_Group_Code"
            ].read(
                1,
                window=window,
            )[valid]

            lith = np.rint(
                lith
            ).astype(
                np.int16,
                copy=False,
            )


            for j, code in enumerate(
                LITHOLOGY_CODES
            ):

                X[:, 6 + j] = (
                    lith == code
                ).astype(np.float32)


            # ---------------------------------------------------------
            # LULC → 8 ONE-HOT FEATURES
            # ---------------------------------------------------------

            lulc = sources[
                "LULC_WorldCover_Code"
            ].read(
                1,
                window=window,
            )[valid]

            lulc = np.rint(
                lulc
            ).astype(
                np.int16,
                copy=False,
            )


            for j, code in enumerate(
                LULC_CODES
            ):

                X[:, 17 + j] = (
                    lulc == code
                ).astype(np.float32)


            # ---------------------------------------------------------
            # EXACT XGBOOST FEATURE MATRIX
            # ---------------------------------------------------------

            dmatrix = xgb.DMatrix(
                X,
                feature_names=FEATURES,
            )


            predictions = np.empty(
                (5, n_valid),
                dtype=np.float32,
            )


            for model_index, booster in enumerate(
                boosters
            ):

                predictions[
                    model_index
                ] = booster.predict(
                    dmatrix
                ).astype(
                    np.float32,
                    copy=False,
                )


            mean_prediction = np.mean(
                predictions,
                axis=0,
                dtype=np.float64,
            ).astype(np.float32)


            sd_prediction = np.std(
                predictions,
                axis=0,
                dtype=np.float64,
            ).astype(np.float32)


            mean_tile[valid] = mean_prediction
            sd_tile[valid] = sd_prediction


            processed_valid_cells += n_valid


            susceptibility_sum += float(
                np.sum(
                    mean_prediction,
                    dtype=np.float64,
                )
            )

            sd_sum += float(
                np.sum(
                    sd_prediction,
                    dtype=np.float64,
                )
            )


            global_mean_min = min(
                global_mean_min,
                float(mean_prediction.min()),
            )

            global_mean_max = max(
                global_mean_max,
                float(mean_prediction.max()),
            )

            global_sd_min = min(
                global_sd_min,
                float(sd_prediction.min()),
            )

            global_sd_max = max(
                global_sd_max,
                float(sd_prediction.max()),
            )


        mean_dst.write(
            mean_tile,
            1,
            window=window,
        )

        sd_dst.write(
            sd_tile,
            1,
            window=window,
        )


        if (
            window_index == 1
            or window_index % 10 == 0
            or window_index == total_windows
        ):

            elapsed = time.time() - start_time

            progress = (
                100.0
                * window_index
                / total_windows
            )

            print(
                f"Window "
                f"{window_index:03d}/"
                f"{total_windows:03d} "
                f"| {progress:6.2f}% "
                f"| Valid cells processed: "
                f"{processed_valid_cells:,} "
                f"| Elapsed: "
                f"{elapsed / 60:.1f} min"
            )


    mean_dst.update_tags(
        model="XGBoost Baseline-v1",
        ensemble_models="5",
        output="Mean susceptibility score",
        rainfall="CHELSA BIO12 1981-2010",
        interpretation=(
            "Relative landslide susceptibility score; "
            "not calibrated absolute event probability"
        ),
    )


    sd_dst.update_tags(
        model="XGBoost Baseline-v1",
        ensemble_models="5",
        output="Standard deviation across five background-realization models",
        interpretation=(
            "Model disagreement/stability indicator; "
            "not complete predictive uncertainty"
        ),
    )


elapsed_seconds = time.time() - start_time


mean_of_susceptibility = (
    susceptibility_sum
    / processed_valid_cells
)

mean_of_sd = (
    sd_sum
    / processed_valid_cells
)


manifest = {
    "model": "Manipur XGBoost Baseline-v1",
    "ensemble_models": 5,
    "feature_count": len(FEATURES),
    "features": FEATURES,
    "lithology_codes": LITHOLOGY_CODES,
    "lulc_codes": LULC_CODES,
    "valid_cells": processed_valid_cells,
    "expected_valid_cells": EXPECTED_VALID_CELLS,
    "valid_cell_count_match": (
        processed_valid_cells
        == EXPECTED_VALID_CELLS
    ),
    "grid_shape": [
        6901,
        6025,
    ],
    "window_size": WINDOW_SIZE,
    "mean_susceptibility_min": global_mean_min,
    "mean_susceptibility_max": global_mean_max,
    "mean_susceptibility_average": mean_of_susceptibility,
    "prediction_sd_min": global_sd_min,
    "prediction_sd_max": global_sd_max,
    "prediction_sd_average": mean_of_sd,
    "elapsed_seconds": elapsed_seconds,
    "mean_raster": MEAN_PATH.name,
    "sd_raster": SD_PATH.name,
    "models": [
        p.name for p in model_paths
    ],
    "rasters": {
        key: value.name
        for key, value
        in raster_paths.items()
    },
}


with open(
    MANIFEST_PATH,
    "w",
) as f:

    json.dump(
        manifest,
        f,
        indent=2,
    )


print()
print("=" * 100)
print("STATEWIDE PREDICTION COMPLETE ✅")
print("=" * 100)

print(
    "Valid cells predicted :",
    f"{processed_valid_cells:,}",
)

print(
    "Susceptibility min    :",
    f"{global_mean_min:.6f}",
)

print(
    "Susceptibility max    :",
    f"{global_mean_max:.6f}",
)

print(
    "Susceptibility mean   :",
    f"{mean_of_susceptibility:.6f}",
)

print(
    "Prediction SD min     :",
    f"{global_sd_min:.6f}",
)

print(
    "Prediction SD max     :",
    f"{global_sd_max:.6f}",
)

print(
    "Mean prediction SD    :",
    f"{mean_of_sd:.6f}",
)

print(
    "Elapsed minutes       :",
    f"{elapsed_seconds / 60:.2f}",
)

print()
print("Mean susceptibility raster:")
print(MEAN_PATH)

print()
print("Prediction SD raster:")
print(SD_PATH)

print()
print("Manifest:")
print(MANIFEST_PATH)

print()
print(
    "IMPORTANT: susceptibility values are "
    "relative susceptibility scores, not "
    "calibrated absolute landslide probabilities."
)
```

```python
%%writefile scripts/91_modeling/check_baseline_v1_overfitting.py

from pathlib import Path
import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    roc_auc_score,
    average_precision_score,
    log_loss,
    brier_score_loss,
    confusion_matrix,
)

ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = ROOT / "output" / "baseline_v1_model_matrices"

OUT_DIR = ROOT / "output" / "baseline_v1_overfitting_check"
OUT_DIR.mkdir(parents=True, exist_ok=True)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",
    "Distance_to_Road_m",
    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",
    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


PARAMS = {
    "objective": "binary:logistic",
    "n_estimators": 300,
    "learning_rate": 0.05,
    "max_depth": 4,
    "min_child_weight": 1.0,
    "gamma": 0.0,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "reg_alpha": 0.0,
    "reg_lambda": 1.0,
    "scale_pos_weight": 1.0,
    "tree_method": "hist",
    "device": "cpu",
    "eval_metric": "logloss",
    "random_state": 20260816,
    "n_jobs": -1,
    "verbosity": 0,
}


def metrics(y, p):
    pred = (p >= 0.5).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y, pred, labels=[0, 1]
    ).ravel()

    specificity = tn / (tn + fp)

    return {
        "ROC_AUC": roc_auc_score(y, p),
        "Average_Precision": average_precision_score(y, p),
        "Accuracy": accuracy_score(y, pred),
        "Balanced_Accuracy": balanced_accuracy_score(y, pred),
        "Precision": precision_score(y, pred, zero_division=0),
        "Recall": recall_score(y, pred, zero_division=0),
        "Specificity": specificity,
        "F1": f1_score(y, pred, zero_division=0),
        "MCC": matthews_corrcoef(y, pred),
        "Log_Loss": log_loss(y, p),
        "Brier": brier_score_loss(y, p),
    }


rows = []

print("=" * 90)
print("BASELINE V1 — TRAIN VS SPATIAL-TEST DIAGNOSTIC")
print("=" * 90)


for set_id in range(1, 6):

    path = MATRIX_DIR / (
        f"manipur_baseline_v1_model_matrix_set_{set_id:02d}.csv"
    )

    df = pd.read_csv(path)

    for fold in range(1, 6):

        train = df[df["CV_Fold"] != fold]
        test = df[df["CV_Fold"] == fold]

        X_train = train[FEATURES]
        y_train = train["Label"].astype(int)

        X_test = test[FEATURES]
        y_test = test["Label"].astype(int)

        model = xgb.XGBClassifier(**PARAMS)
        model.fit(X_train, y_train)

        p_train = model.predict_proba(X_train)[:, 1]
        p_test = model.predict_proba(X_test)[:, 1]

        train_m = metrics(y_train, p_train)
        test_m = metrics(y_test, p_test)

        row = {
            "Dataset_Set": set_id,
            "Fold": fold,
        }

        for key in train_m:
            row[f"Train_{key}"] = train_m[key]
            row[f"Test_{key}"] = test_m[key]

        row["AUC_Gap"] = (
            train_m["ROC_AUC"] - test_m["ROC_AUC"]
        )

        row["F1_Gap"] = (
            train_m["F1"] - test_m["F1"]
        )

        rows.append(row)

        print(
            f"Set {set_id:02d} Fold {fold} | "
            f"Train AUC={train_m['ROC_AUC']:.4f} | "
            f"Spatial Test AUC={test_m['ROC_AUC']:.4f} | "
            f"Gap={row['AUC_Gap']:.4f}"
        )


results = pd.DataFrame(rows)

results.to_csv(
    OUT_DIR / "train_vs_spatial_test_metrics.csv",
    index=False,
)


print()
print("=" * 90)
print("OVERALL TRAIN VS SPATIAL TEST")
print("=" * 90)

for metric in [
    "ROC_AUC",
    "Average_Precision",
    "Accuracy",
    "Balanced_Accuracy",
    "Precision",
    "Recall",
    "Specificity",
    "F1",
    "MCC",
    "Log_Loss",
    "Brier",
]:

    train_mean = results[f"Train_{metric}"].mean()
    test_mean = results[f"Test_{metric}"].mean()

    print(
        f"{metric:20s} | "
        f"Train={train_mean:.4f} | "
        f"Spatial Test={test_mean:.4f}"
    )


print()
print("Mean AUC gap :", results["AUC_Gap"].mean())
print("Max AUC gap  :", results["AUC_Gap"].max())

print("Mean F1 gap  :", results["F1_Gap"].mean())
print("Max F1 gap   :", results["F1_Gap"].max())

print()
print("Saved:")
print(OUT_DIR / "train_vs_spatial_test_metrics.csv")
```

```python
%%writefile scripts/91_modeling/analyze_baseline_v1_shap.py

from pathlib import Path
import json

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = ROOT / "output" / "baseline_v1_model_matrices"

MODEL_DIR = (
    ROOT
    / "output"
    / "xgboost_baseline_v1_final_ensemble"
)

OUT_DIR = (
    ROOT
    / "output"
    / "baseline_v1_shap"
)

OUT_DIR.mkdir(parents=True, exist_ok=True)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",
    "Distance_to_Road_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


# -------------------------------------------------------------------------
# Physical factor grouping
# -------------------------------------------------------------------------

FEATURE_GROUPS = {
    "Elevation": [
        "Elevation_m"
    ],

    "Slope": [
        "Slope_deg"
    ],

    "TWI": [
        "TWI"
    ],

    "Rainfall": [
        "Rainfall_CHELSA_Annual_mm"
    ],

    "Distance to River": [
        "Distance_to_River_m"
    ],

    "Distance to Road": [
        "Distance_to_Road_m"
    ],

    "Lithology": [
        f"Lithology_Group_{i}"
        for i in range(1, 12)
    ],

    "LULC": [
        "LULC_WorldCover_10",
        "LULC_WorldCover_20",
        "LULC_WorldCover_30",
        "LULC_WorldCover_40",
        "LULC_WorldCover_50",
        "LULC_WorldCover_60",
        "LULC_WorldCover_80",
        "LULC_WorldCover_90",
    ],
}


print("=" * 100)
print("MANIPUR XGBOOST BASELINE V1 — SHAP ANALYSIS")
print("=" * 100)

print("Final models      : 5")
print("Frozen features   :", len(FEATURES))
print("Physical factors  :", len(FEATURE_GROUPS))
print()


all_model_importance = []
all_signed_importance = []


for set_id in range(1, 6):

    print("-" * 100)
    print(f"MODEL {set_id}/5")
    print("-" * 100)

    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_model_matrix_set_{set_id:02d}.csv"
    )

    model_path = (
        MODEL_DIR
        / f"manipur_xgb_baseline_v1_final_set_{set_id:02d}.json"
    )

    df = pd.read_csv(matrix_path)

    X = df[FEATURES].astype(np.float32)

    booster = xgb.Booster()
    booster.load_model(str(model_path))

    dmatrix = xgb.DMatrix(
        X,
        feature_names=FEATURES,
    )

    # XGBoost TreeSHAP:
    # output = 25 feature contributions + 1 bias term
    contributions = booster.predict(
        dmatrix,
        pred_contribs=True,
    )

    shap_values = contributions[:, :-1]
    base_values = contributions[:, -1]

    if shap_values.shape[1] != len(FEATURES):
        raise RuntimeError(
            f"Unexpected SHAP feature count for Set {set_id:02d}"
        )

    mean_abs = np.mean(
        np.abs(shap_values),
        axis=0,
    )

    mean_signed = np.mean(
        shap_values,
        axis=0,
    )

    importance_df = pd.DataFrame({
        "Dataset_Set": set_id,
        "Feature": FEATURES,
        "Mean_Abs_SHAP": mean_abs,
        "Mean_Signed_SHAP": mean_signed,
    })

    importance_df.to_csv(
        OUT_DIR /
        f"shap_feature_importance_set_{set_id:02d}.csv",
        index=False,
    )

    all_model_importance.append(mean_abs)
    all_signed_importance.append(mean_signed)

    top = (
        importance_df
        .sort_values(
            "Mean_Abs_SHAP",
            ascending=False
        )
        .head(10)
    )

    print("Top 10 features:")

    for _, row in top.iterrows():

        print(
            f"{row['Feature']:32s} "
            f"| Mean |SHAP| = "
            f"{row['Mean_Abs_SHAP']:.6f}"
        )

    print(
        "Mean base value:",
        float(np.mean(base_values)),
    )


# =============================================================================
# AGGREGATE ACROSS FIVE MODELS
# =============================================================================

importance_matrix = np.vstack(
    all_model_importance
)

signed_matrix = np.vstack(
    all_signed_importance
)


mean_importance = importance_matrix.mean(axis=0)
sd_importance = importance_matrix.std(axis=0)

mean_signed = signed_matrix.mean(axis=0)


global_df = pd.DataFrame({
    "Feature": FEATURES,
    "Mean_Abs_SHAP_5_Models": mean_importance,
    "SD_Abs_SHAP_5_Models": sd_importance,
    "Mean_Signed_SHAP_5_Models": mean_signed,
})


global_df = global_df.sort_values(
    "Mean_Abs_SHAP_5_Models",
    ascending=False,
)


global_df["Rank"] = np.arange(
    1,
    len(global_df) + 1,
)


global_df.to_csv(
    OUT_DIR /
    "shap_global_feature_importance_25_features.csv",
    index=False,
)


# =============================================================================
# AGGREGATE ONE-HOT FEATURES TO 8 PHYSICAL FACTORS
# =============================================================================

feature_to_index = {
    feature: i
    for i, feature in enumerate(FEATURES)
}


group_rows = []


for group_name, group_features in FEATURE_GROUPS.items():

    indices = [
        feature_to_index[f]
        for f in group_features
    ]

    # For grouped categorical variables, sum absolute contributions
    # across their one-hot columns at row/model-summary level.
    group_per_model = (
        importance_matrix[:, indices].sum(axis=1)
    )

    group_signed_per_model = (
        signed_matrix[:, indices].sum(axis=1)
    )

    group_rows.append({
        "Physical_Factor": group_name,
        "Mean_Abs_SHAP_5_Models":
            float(group_per_model.mean()),

        "SD_Abs_SHAP_5_Models":
            float(group_per_model.std()),

        "Mean_Signed_SHAP_5_Models":
            float(group_signed_per_model.mean()),
    })


group_df = pd.DataFrame(group_rows)


group_df = group_df.sort_values(
    "Mean_Abs_SHAP_5_Models",
    ascending=False,
)


group_df["Rank"] = np.arange(
    1,
    len(group_df) + 1,
)


group_df.to_csv(
    OUT_DIR /
    "shap_global_importance_8_physical_factors.csv",
    index=False,
)


# =============================================================================
# PRINT FINAL RESULTS
# =============================================================================

print()
print("=" * 100)
print("GLOBAL SHAP IMPORTANCE — 25 MODEL FEATURES")
print("=" * 100)

print(
    global_df[
        [
            "Rank",
            "Feature",
            "Mean_Abs_SHAP_5_Models",
            "SD_Abs_SHAP_5_Models",
        ]
    ].to_string(index=False)
)


print()
print("=" * 100)
print("GLOBAL SHAP IMPORTANCE — 8 PHYSICAL CONDITIONING FACTORS")
print("=" * 100)

print(
    group_df[
        [
            "Rank",
            "Physical_Factor",
            "Mean_Abs_SHAP_5_Models",
            "SD_Abs_SHAP_5_Models",
        ]
    ].to_string(index=False)
)


print()
print("=" * 100)
print("SHAP ANALYSIS COMPLETE ✅")
print("=" * 100)

print("Output directory:")
print(OUT_DIR)

print()
print(
    "IMPORTANT: Mean absolute SHAP measures influence, "
    "not causality."
)
```

```python
%%writefile /content/Manipur_XGBoost_Colab/scripts/91_modeling/shap_dependence_distance_to_road.py

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = ROOT / "output" / "baseline_v1_model_matrices"

MODEL_DIR = (
    ROOT
    / "output"
    / "xgboost_baseline_v1_final_ensemble"
)

OUT_DIR = (
    ROOT
    / "output"
    / "baseline_v1_shap"
    / "dependence"
)

OUT_DIR.mkdir(parents=True, exist_ok=True)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",
    "Distance_to_Road_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


ROAD_FEATURE = "Distance_to_Road_m"
ROAD_INDEX = FEATURES.index(ROAD_FEATURE)


print("=" * 100)
print("BASELINE V1 — SHAP DEPENDENCE")
print("FEATURE: DISTANCE TO ROAD")
print("=" * 100)


all_rows = []


for set_id in range(1, 6):

    print(f"Processing model {set_id}/5...")

    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_model_matrix_set_{set_id:02d}.csv"
    )

    model_path = (
        MODEL_DIR
        / f"manipur_xgb_baseline_v1_final_set_{set_id:02d}.json"
    )

    df = pd.read_csv(matrix_path)

    X = df[FEATURES].astype(np.float32)

    booster = xgb.Booster()
    booster.load_model(str(model_path))

    dmatrix = xgb.DMatrix(
        X,
        feature_names=FEATURES,
    )

    contributions = booster.predict(
        dmatrix,
        pred_contribs=True,
    )

    road_shap = contributions[:, ROAD_INDEX]

    temp = pd.DataFrame({
        "Dataset_Set": set_id,
        "Distance_to_Road_m": X[ROAD_FEATURE].to_numpy(),
        "SHAP_Distance_to_Road": road_shap,
        "Label": df["Label"].astype(int).to_numpy(),
    })

    all_rows.append(temp)


combined = pd.concat(
    all_rows,
    ignore_index=True,
)


combined.to_csv(
    OUT_DIR / "distance_to_road_shap_values.csv",
    index=False,
)


# ---------------------------------------------------------------------
# ROAD-DISTANCE BANDS
# ---------------------------------------------------------------------

bins = [
    0,
    50,
    100,
    250,
    500,
    1000,
    2000,
    5000,
    10000,
    np.inf,
]

labels = [
    "0-50 m",
    "50-100 m",
    "100-250 m",
    "250-500 m",
    "500-1000 m",
    "1000-2000 m",
    "2000-5000 m",
    "5000-10000 m",
    ">10000 m",
]


combined["Road_Distance_Band"] = pd.cut(
    combined["Distance_to_Road_m"],
    bins=bins,
    labels=labels,
    include_lowest=True,
    right=False,
)


summary = (
    combined
    .groupby(
        "Road_Distance_Band",
        observed=True,
    )
    .agg(
        Samples=("Distance_to_Road_m", "size"),
        Median_Distance_m=("Distance_to_Road_m", "median"),
        Mean_SHAP=("SHAP_Distance_to_Road", "mean"),
        Median_SHAP=("SHAP_Distance_to_Road", "median"),
        Mean_Abs_SHAP=(
            "SHAP_Distance_to_Road",
            lambda x: np.mean(np.abs(x))
        ),
        Positive_Rate=("Label", "mean"),
    )
    .reset_index()
)


summary.to_csv(
    OUT_DIR / "distance_to_road_shap_binned_summary.csv",
    index=False,
)


# ---------------------------------------------------------------------
# POSITIVE VS BACKGROUND ROAD DISTANCE
# ---------------------------------------------------------------------

class_summary = (
    combined
    .groupby("Label")
    .agg(
        Samples=("Distance_to_Road_m", "size"),
        Mean_Distance_m=("Distance_to_Road_m", "mean"),
        Median_Distance_m=("Distance_to_Road_m", "median"),
        Mean_SHAP=("SHAP_Distance_to_Road", "mean"),
        Median_SHAP=("SHAP_Distance_to_Road", "median"),
    )
    .reset_index()
)


class_summary["Class"] = class_summary["Label"].map({
    0: "Background",
    1: "Landslide",
})


class_summary.to_csv(
    OUT_DIR / "distance_to_road_class_summary.csv",
    index=False,
)


# ---------------------------------------------------------------------
# CORRELATIONS
# ---------------------------------------------------------------------

spearman = combined[
    ["Distance_to_Road_m", "SHAP_Distance_to_Road"]
].corr(method="spearman").iloc[0, 1]


pearson = combined[
    ["Distance_to_Road_m", "SHAP_Distance_to_Road"]
].corr(method="pearson").iloc[0, 1]


# ---------------------------------------------------------------------
# PRINT
# ---------------------------------------------------------------------

print()
print("=" * 100)
print("DISTANCE-TO-ROAD SHAP SUMMARY")
print("=" * 100)

print(summary.to_string(index=False))


print()
print("=" * 100)
print("LANDSLIDE VS BACKGROUND ROAD DISTANCE")
print("=" * 100)

print(
    class_summary[
        [
            "Class",
            "Samples",
            "Mean_Distance_m",
            "Median_Distance_m",
            "Mean_SHAP",
            "Median_SHAP",
        ]
    ].to_string(index=False)
)


print()
print(
    "Spearman correlation (Road Distance vs SHAP):",
    f"{spearman:.4f}",
)

print(
    "Pearson correlation  (Road Distance vs SHAP):",
    f"{pearson:.4f}",
)


print()
print("Interpretation:")
print("SHAP > 0  = pushes model toward HIGHER susceptibility")
print("SHAP < 0  = pushes model toward LOWER susceptibility")


print()
print("=" * 100)
print("DISTANCE-TO-ROAD DEPENDENCE COMPLETE ✅")
print("=" * 100)

print()
print("Saved:")
print(
    OUT_DIR / "distance_to_road_shap_binned_summary.csv"
)
print(
    OUT_DIR / "distance_to_road_class_summary.csv"
)
```

```python
%%writefile scripts/91_modeling/check_road_inventory_bias.py

from pathlib import Path
import numpy as np
import pandas as pd

ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = ROOT / "output" / "baseline_v1_model_matrices"

OUT_DIR = ROOT / "output" / "road_bias_check"
OUT_DIR.mkdir(parents=True, exist_ok=True)


all_sets = []

for set_id in range(1, 6):

    path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_model_matrix_set_{set_id:02d}.csv"
    )

    df = pd.read_csv(path)

    all_sets.append(
        df[
            [
                "Dataset_Set",
                "Cell_ID",
                "Label",
                "Distance_to_Road_m",
            ]
        ]
    )


combined = pd.concat(
    all_sets,
    ignore_index=True,
)


# ------------------------------------------------------------------
# UNIQUE POSITIVE CELLS
# Positives occur in all 5 datasets, so remove duplicates by Cell_ID.
# ------------------------------------------------------------------

positive = (
    combined[
        combined["Label"] == 1
    ]
    .drop_duplicates(
        subset="Cell_ID"
    )
    .copy()
)


# ------------------------------------------------------------------
# UNIQUE BACKGROUND CELLS
# Remove any repeated background cell occurring across realizations.
# ------------------------------------------------------------------

background = (
    combined[
        combined["Label"] == 0
    ]
    .drop_duplicates(
        subset="Cell_ID"
    )
    .copy()
)


def summarize(name, df):

    d = df["Distance_to_Road_m"]

    return {
        "Class": name,
        "Unique_Cells": len(df),
        "Mean_m": d.mean(),
        "Median_m": d.median(),
        "Min_m": d.min(),
        "Max_m": d.max(),

        "At_Road_0m_pct":
            100 * (d <= 1e-6).mean(),

        "Within_50m_pct":
            100 * (d < 50).mean(),

        "Within_100m_pct":
            100 * (d < 100).mean(),

        "Within_250m_pct":
            100 * (d < 250).mean(),

        "Within_500m_pct":
            100 * (d < 500).mean(),

        "Within_1000m_pct":
            100 * (d < 1000).mean(),
    }


summary = pd.DataFrame(
    [
        summarize(
            "Landslide",
            positive
        ),

        summarize(
            "Background",
            background
        ),
    ]
)


print("=" * 100)
print("ROAD-DISTANCE INVENTORY BIAS CHECK")
print("=" * 100)

print()
print(summary.to_string(index=False))


print()
print("=" * 100)
print("KEY CHECK")
print("=" * 100)

print(
    "Unique landslide cells :",
    f"{len(positive):,}"
)

print(
    "Unique background cells:",
    f"{len(background):,}"
)

print()

print(
    "Landslides exactly on road (%):",
    f"{100 * (positive['Distance_to_Road_m'] <= 1e-6).mean():.2f}%"
)

print(
    "Landslides within 50 m (%):",
    f"{100 * (positive['Distance_to_Road_m'] < 50).mean():.2f}%"
)

print(
    "Background within 50 m (%):",
    f"{100 * (background['Distance_to_Road_m'] < 50).mean():.2f}%"
)


summary.to_csv(
    OUT_DIR / "road_distance_unique_cell_summary.csv",
    index=False,
)

positive.to_csv(
    OUT_DIR / "unique_positive_road_distances.csv",
    index=False,
)

background.to_csv(
    OUT_DIR / "unique_background_road_distances.csv",
    index=False,
)


print()
print("ROAD-BIAS CHECK COMPLETE ✅")
```

```python
%%writefile scripts/91_modeling/run_baseline_v1_no_road_ablation.py

from pathlib import Path
import json

import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    log_loss,
    brier_score_loss,
    confusion_matrix,
)


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = ROOT / "output" / "baseline_v1_model_matrices"

OUT_DIR = ROOT / "output" / "baseline_v1_no_road_ablation"
OUT_DIR.mkdir(parents=True, exist_ok=True)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    # Distance_to_Road_m deliberately removed

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


PARAMS = {
    "objective": "binary:logistic",
    "n_estimators": 300,
    "learning_rate": 0.05,
    "max_depth": 4,
    "min_child_weight": 1.0,
    "gamma": 0.0,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "reg_alpha": 0.0,
    "reg_lambda": 1.0,
    "scale_pos_weight": 1.0,
    "tree_method": "hist",
    "device": "cpu",
    "eval_metric": "logloss",
    "random_state": 20260816,
    "n_jobs": -1,
    "verbosity": 0,
}


def calculate_metrics(y_true, probabilities):

    pred = (probabilities >= 0.5).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        pred,
        labels=[0, 1],
    ).ravel()

    specificity = tn / (tn + fp)

    return {
        "ROC_AUC":
            roc_auc_score(y_true, probabilities),

        "Average_Precision":
            average_precision_score(
                y_true,
                probabilities
            ),

        "Accuracy_0_5":
            accuracy_score(y_true, pred),

        "Balanced_Accuracy_0_5":
            balanced_accuracy_score(
                y_true,
                pred
            ),

        "Precision_0_5":
            precision_score(
                y_true,
                pred,
                zero_division=0
            ),

        "Recall_0_5":
            recall_score(
                y_true,
                pred,
                zero_division=0
            ),

        "Specificity_0_5":
            specificity,

        "F1_0_5":
            f1_score(
                y_true,
                pred,
                zero_division=0
            ),

        "MCC_0_5":
            matthews_corrcoef(
                y_true,
                pred
            ),

        "Log_Loss":
            log_loss(
                y_true,
                probabilities
            ),

        "Brier_Score":
            brier_score_loss(
                y_true,
                probabilities
            ),
    }


print("=" * 100)
print("BASELINE V1 — ROAD ABLATION EXPERIMENT")
print("=" * 100)

print("Original features : 25")
print("Removed feature   : Distance_to_Road_m")
print("Remaining features:", len(FEATURES))
print("Spatial CV        : 5 sets × 5 folds")
print()


fold_rows = []
dataset_rows = []


for set_id in range(1, 6):

    path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_model_matrix_set_{set_id:02d}.csv"
    )

    df = pd.read_csv(path)

    print()
    print("-" * 100)
    print(f"DATASET SET {set_id:02d}")
    print("-" * 100)

    oof_probabilities = np.full(
        len(df),
        np.nan,
        dtype=float,
    )


    for fold in range(1, 6):

        train_mask = df["CV_Fold"] != fold
        test_mask = df["CV_Fold"] == fold

        train = df.loc[train_mask]
        test = df.loc[test_mask]

        X_train = train[FEATURES]
        y_train = train["Label"].astype(int)

        X_test = test[FEATURES]
        y_test = test["Label"].astype(int)

        model = xgb.XGBClassifier(**PARAMS)

        model.fit(
            X_train,
            y_train
        )

        probabilities = model.predict_proba(
            X_test
        )[:, 1]

        oof_probabilities[test_mask.to_numpy()] = probabilities

        metrics = calculate_metrics(
            y_test,
            probabilities
        )

        metrics.update({
            "Dataset_Set": set_id,
            "Fold": fold,
        })

        fold_rows.append(metrics)

        print(
            f"Fold {fold} | "
            f"ROC-AUC={metrics['ROC_AUC']:.6f} | "
            f"AP={metrics['Average_Precision']:.6f} | "
            f"F1={metrics['F1_0_5']:.6f}"
        )


    if np.isnan(oof_probabilities).any():
        raise RuntimeError(
            f"Missing OOF predictions in Set {set_id}"
        )

    dataset_metrics = calculate_metrics(
        df["Label"].astype(int),
        oof_probabilities
    )

    dataset_metrics["Dataset_Set"] = set_id

    dataset_rows.append(dataset_metrics)

    print()
    print(
        f"SET {set_id:02d} OOF | "
        f"ROC-AUC={dataset_metrics['ROC_AUC']:.6f} | "
        f"AP={dataset_metrics['Average_Precision']:.6f} | "
        f"F1={dataset_metrics['F1_0_5']:.6f}"
    )


fold_df = pd.DataFrame(fold_rows)

dataset_df = pd.DataFrame(dataset_rows)


fold_df.to_csv(
    OUT_DIR / "no_road_fold_metrics.csv",
    index=False,
)

dataset_df.to_csv(
    OUT_DIR / "no_road_dataset_oof_metrics.csv",
    index=False,
)


metrics_to_summarize = [
    "ROC_AUC",
    "Average_Precision",
    "Accuracy_0_5",
    "Balanced_Accuracy_0_5",
    "Precision_0_5",
    "Recall_0_5",
    "Specificity_0_5",
    "F1_0_5",
    "MCC_0_5",
    "Log_Loss",
    "Brier_Score",
]


summary_rows = []


for metric in metrics_to_summarize:

    values = dataset_df[metric]

    summary_rows.append({
        "Metric": metric,
        "Mean": values.mean(),
        "SD": values.std(ddof=1),
        "Min": values.min(),
        "Max": values.max(),
    })


summary_df = pd.DataFrame(
    summary_rows
)


summary_df.to_csv(
    OUT_DIR / "no_road_primary_summary.csv",
    index=False,
)


print()
print("=" * 100)
print("NO-ROAD MODEL — FINAL SUMMARY")
print("=" * 100)

print(
    summary_df.to_string(
        index=False
    )
)


print()
print("=" * 100)
print("ROAD ABLATION COMPLETE ✅")
print("=" * 100)

print("Output:")
print(OUT_DIR)
```

```python
%%writefile scripts/91_modeling/preflight_road_matched_background.py

from pathlib import Path
import numpy as np
import pandas as pd
import rasterio
from rasterio.windows import Window


ROOT = Path("/content/Manipur_XGBoost_Colab")
RASTER_DIR = Path("/content/Manipur_XGBoost_Statewide_Rasters")

MATRIX_PATH = (
    ROOT
    / "output"
    / "baseline_v1_model_matrices"
    / "manipur_baseline_v1_model_matrix_set_01.csv"
)

BLOCK_MAP_PATH = (
    ROOT
    / "output"
    / "spatial_cv_10km"
    / "manipur_10km_block_to_cv_fold.csv"
)

MASK_PATH = (
    RASTER_DIR
    / "manipur_baseline_common_valid_mask_30m.tif"
)

ROAD_PATH = (
    RASTER_DIR
    / "manipur_distance_to_road_30m.tif"
)


# Exact matched-background targets
TARGETS = [
    ("0-50 m",        0,       50,       2364),
    ("50-100 m",      50,      100,       111),
    ("100-250 m",     100,     250,       171),
    ("250-500 m",     250,     500,       120),
    ("500-1000 m",    500,     1000,      135),
    ("1000-2000 m",   1000,    2000,      159),
    ("2000-5000 m",   2000,    5000,      180),
    ("5000-10000 m",  5000,    10000,     165),
    (">10000 m",      10000,   np.inf,       3),
]


WINDOW_SIZE = 512
BLOCK_SIZE_M = 10000.0


print("=" * 100)
print("ROAD-DISTANCE-MATCHED BACKGROUND — PREFLIGHT")
print("=" * 100)


# =============================================================================
# POSITIVE CELLS
# =============================================================================

df = pd.read_csv(MATRIX_PATH)

positive = (
    df[df["Label"] == 1]
    .drop_duplicates("Cell_ID")
    .copy()
)

if len(positive) != 1136:
    raise RuntimeError(
        f"Expected 1,136 unique positives, found {len(positive)}"
    )

positive_rows = positive["Raster_Row"].astype(np.int64).to_numpy()
positive_cols = positive["Raster_Col"].astype(np.int64).to_numpy()


# =============================================================================
# EXISTING FROZEN BLOCK → FOLD MAP
# =============================================================================

block_map = pd.read_csv(BLOCK_MAP_PATH)

required_columns = {
    "Spatial_Block_Row",
    "Spatial_Block_Col",
    "Spatial_Block_ID",
    "CV_Fold",
}

missing = required_columns - set(block_map.columns)

if missing:
    raise RuntimeError(
        f"Block map missing columns: {sorted(missing)}"
    )


mapped_pairs = set(
    zip(
        block_map["Spatial_Block_Row"].astype(int),
        block_map["Spatial_Block_Col"].astype(int),
    )
)


print("Unique positive cells :", f"{len(positive):,}")
print("Mapped 10 km blocks   :", f"{len(mapped_pairs):,}")

print()
print("Existing fold distribution:")

print(
    block_map["CV_Fold"]
    .value_counts()
    .sort_index()
    .to_string()
)


# =============================================================================
# STATEWIDE SCAN
# =============================================================================

candidate_counts = {
    label: 0
    for label, _, _, _ in TARGETS
}

valid_blocks = set()

total_valid = 0
eligible_background = 0


with rasterio.open(MASK_PATH) as mask_src, \
     rasterio.open(ROAD_PATH) as road_src:

    if (
        mask_src.crs != road_src.crs
        or mask_src.shape != road_src.shape
        or mask_src.transform != road_src.transform
    ):
        raise RuntimeError(
            "Mask and road raster grids do not match."
        )

    height = mask_src.height
    width = mask_src.width
    transform = mask_src.transform

    # Positive cells as flattened raster indexes.
    positive_linear = (
        positive_rows * width
        + positive_cols
    )


    for row_off in range(0, height, WINDOW_SIZE):

        h = min(
            WINDOW_SIZE,
            height - row_off
        )

        for col_off in range(
            0,
            width,
            WINDOW_SIZE
        ):

            w = min(
                WINDOW_SIZE,
                width - col_off
            )

            window = Window(
                col_off,
                row_off,
                w,
                h,
            )

            mask = mask_src.read(
                1,
                window=window,
            )

            valid = mask > 0

            if not valid.any():
                continue

            rr, cc = np.where(valid)

            global_rows = (
                rr.astype(np.int64)
                + row_off
            )

            global_cols = (
                cc.astype(np.int64)
                + col_off
            )

            total_valid += len(global_rows)


            # ---------------------------------------------------------
            # CHECK 10 KM BLOCK COVERAGE
            # ---------------------------------------------------------

            x = (
                transform.c
                + (global_cols + 0.5)
                * transform.a
            )

            y = (
                transform.f
                + (global_rows + 0.5)
                * transform.e
            )

            block_cols = np.floor(
                (x - transform.c)
                / BLOCK_SIZE_M
            ).astype(int)

            block_rows = np.floor(
                (transform.f - y)
                / BLOCK_SIZE_M
            ).astype(int)

            valid_blocks.update(
                zip(
                    block_rows.tolist(),
                    block_cols.tolist(),
                )
            )


            # ---------------------------------------------------------
            # REMOVE POSITIVE CELLS
            # ---------------------------------------------------------

            linear = (
                global_rows * width
                + global_cols
            )

            is_positive = np.isin(
                linear,
                positive_linear,
            )

            background_mask = ~is_positive

            if not background_mask.any():
                continue

            eligible_background += int(
                background_mask.sum()
            )


            # ---------------------------------------------------------
            # ROAD-DISTANCE BAND COUNTS
            # ---------------------------------------------------------

            road = road_src.read(
                1,
                window=window,
            )[valid]

            road = road[
                background_mask
            ].astype(
                np.float64,
                copy=False,
            )


            for (
                label,
                lower,
                upper,
                target,
            ) in TARGETS:

                if np.isinf(upper):

                    count = np.sum(
                        road >= lower
                    )

                else:

                    count = np.sum(
                        (road >= lower)
                        & (road < upper)
                    )

                candidate_counts[label] += int(
                    count
                )


# =============================================================================
# BLOCK COVERAGE
# =============================================================================

unmapped_blocks = (
    valid_blocks
    - mapped_pairs
)


print()
print("=" * 100)
print("STATEWIDE BLOCK COVERAGE")
print("=" * 100)

print(
    "Total valid cells          :",
    f"{total_valid:,}"
)

print(
    "Eligible background cells  :",
    f"{eligible_background:,}"
)

print(
    "Valid 10 km blocks         :",
    f"{len(valid_blocks):,}"
)

print(
    "Mapped 10 km blocks        :",
    f"{len(mapped_pairs):,}"
)

print(
    "Unmapped valid blocks      :",
    len(unmapped_blocks)
)


if unmapped_blocks:

    print()
    print("First unmapped blocks:")

    for item in sorted(
        unmapped_blocks
    )[:20]:

        print(item)


# =============================================================================
# ROAD-BAND AVAILABILITY
# =============================================================================

print()
print("=" * 100)
print("ROAD-BAND BACKGROUND AVAILABILITY")
print("=" * 100)

print(
    f"{'Band':18s} "
    f"{'Available':>12s} "
    f"{'Needed':>10s} "
    f"{'Status':>10s}"
)

print("-" * 55)


all_bands_pass = True


for (
    label,
    lower,
    upper,
    target,
) in TARGETS:

    available = candidate_counts[
        label
    ]

    passed = available >= target

    if not passed:
        all_bands_pass = False

    print(
        f"{label:18s} "
        f"{available:12,} "
        f"{target:10,} "
        f"{'PASS ✅' if passed else 'FAIL ❌':>10s}"
    )


print()
print(
    "Required background total :",
    f"{sum(x[3] for x in TARGETS):,}"
)


# =============================================================================
# FINAL
# =============================================================================

print()
print("=" * 100)

if (
    total_valid == 24_718_359
    and len(unmapped_blocks) == 0
    and all_bands_pass
):

    print(
        "ROAD-MATCHED BACKGROUND PREFLIGHT: PASS ✅"
    )

    print(
        "Safe to generate the five "
        "road-distance-matched background sets."
    )

else:

    print(
        "ROAD-MATCHED BACKGROUND PREFLIGHT: "
        "NOT READY ❌"
    )

    if total_valid != 24_718_359:

        print(
            "Reason: valid-cell count mismatch."
        )

    if unmapped_blocks:

        print(
            "Reason: some statewide blocks "
            "do not have frozen CV-fold assignments."
        )

    if not all_bands_pass:

        print(
            "Reason: insufficient eligible cells "
            "in one or more road-distance bands."
        )

print("=" * 100)
```

```python
%%writefile scripts/91_modeling/generate_road_matched_background_sets.py

from pathlib import Path

import numpy as np
import pandas as pd
import rasterio
from rasterio.windows import Window


# =============================================================================
# PATHS
# =============================================================================

ROOT = Path("/content/Manipur_XGBoost_Colab")
RASTER_DIR = Path("/content/Manipur_XGBoost_Statewide_Rasters")

BASE_MATRIX = (
    ROOT
    / "output"
    / "baseline_v1_model_matrices"
    / "manipur_baseline_v1_model_matrix_set_01.csv"
)

BLOCK_MAP_PATH = (
    ROOT
    / "output"
    / "spatial_cv_10km"
    / "manipur_10km_block_to_cv_fold.csv"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_model_matrices"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


RASTERS = {
    "Elevation_m":
        RASTER_DIR / "manipur_dem_30m_utm46n.tif",

    "Slope_deg":
        RASTER_DIR / "manipur_slope_30m.tif",

    "TWI":
        RASTER_DIR / "manipur_twi_30m.tif",

    "Rainfall_CHELSA_Annual_mm":
        RASTER_DIR / "manipur_rainfall_annual_30m.tif",

    "Distance_to_River_m":
        RASTER_DIR / "manipur_distance_to_river_30m.tif",

    "Distance_to_Road_m":
        RASTER_DIR / "manipur_distance_to_road_30m.tif",

    "Lithology_Code":
        RASTER_DIR / "manipur_lithology_grouped_30m.tif",

    "LULC_Code":
        RASTER_DIR / "manipur_lulc_worldcover_2021_30m.tif",
}


MASK_PATH = (
    RASTER_DIR
    / "manipur_baseline_common_valid_mask_30m.tif"
)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",
    "Distance_to_Road_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


SEEDS = [
    20260815,
    20260816,
    20260817,
    20260818,
    20260819,
]


WINDOW_SIZE = 512
BLOCK_SIZE_M = 10000.0


# =============================================================================
# ROAD STRATIFICATION
# =============================================================================

def road_stratum(d):

    d = np.asarray(d)

    out = np.empty(
        d.shape,
        dtype=object,
    )

    out[d <= 1e-6] = "Exactly_0m"

    out[
        (d > 1e-6)
        & (d < 50)
    ] = "GT0_50m"

    out[
        (d >= 50)
        & (d < 100)
    ] = "50_100m"

    out[
        (d >= 100)
        & (d < 250)
    ] = "100_250m"

    out[
        (d >= 250)
        & (d < 500)
    ] = "250_500m"

    out[
        (d >= 500)
        & (d < 1000)
    ] = "500_1000m"

    out[
        (d >= 1000)
        & (d < 2000)
    ] = "1000_2000m"

    out[
        (d >= 2000)
        & (d < 5000)
    ] = "2000_5000m"

    out[
        (d >= 5000)
        & (d < 10000)
    ] = "5000_10000m"

    out[d >= 10000] = "GE10000m"

    return out


STRATA_ORDER = [
    "Exactly_0m",
    "GT0_50m",
    "50_100m",
    "100_250m",
    "250_500m",
    "500_1000m",
    "1000_2000m",
    "2000_5000m",
    "5000_10000m",
    "GE10000m",
]


# =============================================================================
# LOAD POSITIVES
# =============================================================================

print("=" * 105)
print("ROAD-DISTANCE-MATCHED BACKGROUND GENERATION")
print("=" * 105)


base_df = pd.read_csv(
    BASE_MATRIX
)

positive = (
    base_df[
        base_df["Label"] == 1
    ]
    .drop_duplicates("Cell_ID")
    .copy()
)


if len(positive) != 1136:

    raise RuntimeError(
        f"Expected 1,136 positives, found {len(positive)}"
    )


positive[
    "Road_Match_Stratum"
] = road_stratum(
    positive[
        "Distance_to_Road_m"
    ].to_numpy()
)


positive_counts = (
    positive[
        "Road_Match_Stratum"
    ]
    .value_counts()
    .reindex(
        STRATA_ORDER,
        fill_value=0,
    )
)


targets = (
    positive_counts * 3
).astype(int)


print()
print("POSITIVE ROAD-DISTANCE DISTRIBUTION")
print("-" * 105)

print(
    f"{'Stratum':20s}"
    f"{'Positive':>12s}"
    f"{'Background target':>20s}"
)


for s in STRATA_ORDER:

    print(
        f"{s:20s}"
        f"{positive_counts[s]:12,d}"
        f"{targets[s]:20,d}"
    )


print()
print(
    "Positive total  :",
    f"{positive_counts.sum():,}"
)

print(
    "Background total:",
    f"{targets.sum():,}"
)


if targets.sum() != 3408:

    raise RuntimeError(
        "Expected 3,408 matched backgrounds."
    )


# =============================================================================
# POSITIVE CELL INDEXES
# =============================================================================

positive_rows = (
    positive[
        "Raster_Row"
    ]
    .astype(np.int64)
    .to_numpy()
)

positive_cols = (
    positive[
        "Raster_Col"
    ]
    .astype(np.int64)
    .to_numpy()
)


# =============================================================================
# FROZEN BLOCK → FOLD MAP
# =============================================================================

block_map = pd.read_csv(
    BLOCK_MAP_PATH
)


fold_lookup = {
    (
        int(row.Spatial_Block_Row),
        int(row.Spatial_Block_Col),
    ):
        int(row.CV_Fold)

    for row in block_map.itertuples()
}


mapped_pairs = set(
    fold_lookup.keys()
)


# =============================================================================
# OPEN RASTERS
# =============================================================================

sources = {
    name: rasterio.open(path)
    for name, path in RASTERS.items()
}

mask_src = rasterio.open(
    MASK_PATH
)


try:

    reference = mask_src

    height = reference.height
    width = reference.width
    transform = reference.transform


    for name, src in sources.items():

        if (
            src.crs != reference.crs
            or src.shape != reference.shape
            or src.transform != reference.transform
        ):

            raise RuntimeError(
                f"Raster-grid mismatch: {name}"
            )


    positive_linear = (
        positive_rows * width
        + positive_cols
    )


    # =========================================================================
    # PASS 1 — COUNT AVAILABLE CELLS IN EACH MATCHING STRATUM
    # =========================================================================

    print()
    print("=" * 105)
    print("PASS 1 — COUNTING ELIGIBLE BACKGROUND CELLS")
    print("=" * 105)


    available = {
        s: 0
        for s in STRATA_ORDER
    }


    for row_off in range(
        0,
        height,
        WINDOW_SIZE,
    ):

        h = min(
            WINDOW_SIZE,
            height - row_off,
        )


        for col_off in range(
            0,
            width,
            WINDOW_SIZE,
        ):

            w = min(
                WINDOW_SIZE,
                width - col_off,
            )

            window = Window(
                col_off,
                row_off,
                w,
                h,
            )


            mask = mask_src.read(
                1,
                window=window,
            ) > 0


            if not mask.any():
                continue


            rr, cc = np.where(mask)

            global_rows = (
                rr.astype(np.int64)
                + row_off
            )

            global_cols = (
                cc.astype(np.int64)
                + col_off
            )


            # ---------------------------------------------------------
            # Existing mapped CV blocks only
            # ---------------------------------------------------------

            x = (
                transform.c
                + (global_cols + 0.5)
                * transform.a
            )

            y = (
                transform.f
                + (global_rows + 0.5)
                * transform.e
            )


            block_cols = np.floor(
                (x - transform.c)
                / BLOCK_SIZE_M
            ).astype(int)

            block_rows = np.floor(
                (transform.f - y)
                / BLOCK_SIZE_M
            ).astype(int)


            mapped = np.array(
                [
                    (
                        int(br),
                        int(bc),
                    )
                    in mapped_pairs

                    for br, bc in zip(
                        block_rows,
                        block_cols,
                    )
                ],
                dtype=bool,
            )


            if not mapped.any():
                continue


            global_rows = global_rows[mapped]
            global_cols = global_cols[mapped]


            linear = (
                global_rows * width
                + global_cols
            )


            is_positive = np.isin(
                linear,
                positive_linear,
            )


            eligible = ~is_positive


            if not eligible.any():
                continue


            road_values = (
                sources[
                    "Distance_to_Road_m"
                ]
                .read(
                    1,
                    window=window,
                )[mask][mapped][eligible]
            )


            strata = road_stratum(
                road_values
            )


            for s in STRATA_ORDER:

                available[s] += int(
                    np.sum(
                        strata == s
                    )
                )


    print()
    print(
        f"{'Stratum':20s}"
        f"{'Available':>15s}"
        f"{'Needed/set':>15s}"
        f"{'Status':>12s}"
    )

    print("-" * 65)


    for s in STRATA_ORDER:

        status = (
            "PASS"
            if available[s] >= targets[s]
            else "FAIL"
        )

        print(
            f"{s:20s}"
            f"{available[s]:15,d}"
            f"{targets[s]:15,d}"
            f"{status:>12s}"
        )


        if available[s] < targets[s]:

            raise RuntimeError(
                f"Insufficient candidates in {s}"
            )


    # =========================================================================
    # CHOOSE DETERMINISTIC BAND-LOCAL RANKS
    # =========================================================================

    selected_ranks = {}


    for set_id, seed in enumerate(
        SEEDS,
        start=1,
    ):

        rng = np.random.default_rng(
            seed
        )

        selected_ranks[set_id] = {}


        for s in STRATA_ORDER:

            selected_ranks[
                set_id
            ][s] = np.sort(
                rng.choice(
                    available[s],
                    size=targets[s],
                    replace=False,
                )
            )


    # =========================================================================
    # PASS 2 — COLLECT THE SELECTED CELLS
    # =========================================================================

    print()
    print("=" * 105)
    print("PASS 2 — SAMPLING FIVE MATCHED BACKGROUND REALIZATIONS")
    print("=" * 105)


    seen = {
        s: 0
        for s in STRATA_ORDER
    }


    sampled = {
        set_id: []
        for set_id in range(1, 6)
    }


    for row_off in range(
        0,
        height,
        WINDOW_SIZE,
    ):

        h = min(
            WINDOW_SIZE,
            height - row_off,
        )


        for col_off in range(
            0,
            width,
            WINDOW_SIZE,
        ):

            w = min(
                WINDOW_SIZE,
                width - col_off,
            )

            window = Window(
                col_off,
                row_off,
                w,
                h,
            )


            mask = mask_src.read(
                1,
                window=window,
            ) > 0


            if not mask.any():
                continue


            rr, cc = np.where(
                mask
            )


            global_rows_all = (
                rr.astype(np.int64)
                + row_off
            )

            global_cols_all = (
                cc.astype(np.int64)
                + col_off
            )


            x_all = (
                transform.c
                + (global_cols_all + 0.5)
                * transform.a
            )

            y_all = (
                transform.f
                + (global_rows_all + 0.5)
                * transform.e
            )


            block_cols_all = np.floor(
                (x_all - transform.c)
                / BLOCK_SIZE_M
            ).astype(int)

            block_rows_all = np.floor(
                (transform.f - y_all)
                / BLOCK_SIZE_M
            ).astype(int)


            mapped = np.array(
                [
                    (
                        int(br),
                        int(bc),
                    )
                    in mapped_pairs

                    for br, bc in zip(
                        block_rows_all,
                        block_cols_all,
                    )
                ],
                dtype=bool,
            )


            if not mapped.any():
                continue


            global_rows = global_rows_all[mapped]
            global_cols = global_cols_all[mapped]

            block_rows = block_rows_all[mapped]
            block_cols = block_cols_all[mapped]

            x = x_all[mapped]
            y = y_all[mapped]


            linear = (
                global_rows * width
                + global_cols
            )


            is_positive = np.isin(
                linear,
                positive_linear,
            )


            eligible = ~is_positive


            if not eligible.any():
                continue


            global_rows = global_rows[eligible]
            global_cols = global_cols[eligible]

            block_rows = block_rows[eligible]
            block_cols = block_cols[eligible]

            x = x[eligible]
            y = y[eligible]


            # ---------------------------------------------------------
            # Read all predictor values
            # ---------------------------------------------------------

            values = {}


            for name, src in sources.items():

                values[name] = (
                    src.read(
                        1,
                        window=window,
                    )[mask][mapped][eligible]
                )


            strata = road_stratum(
                values[
                    "Distance_to_Road_m"
                ]
            )


            # ---------------------------------------------------------
            # Process each stratum
            # ---------------------------------------------------------

            for s in STRATA_ORDER:

                band_positions = np.flatnonzero(
                    strata == s
                )


                n_band = len(
                    band_positions
                )


                if n_band == 0:
                    continue


                start = seen[s]
                end = start + n_band


                for set_id in range(
                    1,
                    6,
                ):

                    ranks = (
                        selected_ranks[
                            set_id
                        ][s]
                    )


                    left = np.searchsorted(
                        ranks,
                        start,
                        side="left",
                    )

                    right = np.searchsorted(
                        ranks,
                        end,
                        side="left",
                    )


                    if right <= left:
                        continue


                    local_rank = (
                        ranks[left:right]
                        - start
                    )


                    chosen = band_positions[
                        local_rank
                    ]


                    for idx in chosen:

                        r = int(
                            global_rows[idx]
                        )

                        c = int(
                            global_cols[idx]
                        )

                        br = int(
                            block_rows[idx]
                        )

                        bc = int(
                            block_cols[idx]
                        )


                        lith = int(
                            round(
                                float(
                                    values[
                                        "Lithology_Code"
                                    ][idx]
                                )
                            )
                        )

                        lulc = int(
                            round(
                                float(
                                    values[
                                        "LULC_Code"
                                    ][idx]
                                )
                            )
                        )


                        if lith not in range(
                            1,
                            12,
                        ):

                            raise RuntimeError(
                                f"Unexpected lithology code: {lith}"
                            )


                        if lulc not in {
                            10,
                            20,
                            30,
                            40,
                            50,
                            60,
                            80,
                            90,
                        }:

                            raise RuntimeError(
                                f"Unexpected LULC code: {lulc}"
                            )


                        row = {
                            "Dataset_Set":
                                set_id,

                            "Cell_ID":
                                f"R{r}_C{c}",

                            "Raster_Row":
                                r,

                            "Raster_Col":
                                c,

                            "X_Center":
                                float(x[idx]),

                            "Y_Center":
                                float(y[idx]),

                            "Spatial_Block_Row":
                                br,

                            "Spatial_Block_Col":
                                bc,

                            "Spatial_Block_ID":
                                f"B10KM_R{br}_C{bc}",

                            "CV_Fold":
                                fold_lookup[
                                    (br, bc)
                                ],

                            "Label":
                                0,

                            "Road_Match_Stratum":
                                s,

                            "Elevation_m":
                                float(
                                    values[
                                        "Elevation_m"
                                    ][idx]
                                ),

                            "Slope_deg":
                                float(
                                    values[
                                        "Slope_deg"
                                    ][idx]
                                ),

                            "TWI":
                                float(
                                    values[
                                        "TWI"
                                    ][idx]
                                ),

                            "Rainfall_CHELSA_Annual_mm":
                                float(
                                    values[
                                        "Rainfall_CHELSA_Annual_mm"
                                    ][idx]
                                ),

                            "Distance_to_River_m":
                                float(
                                    values[
                                        "Distance_to_River_m"
                                    ][idx]
                                ),

                            "Distance_to_Road_m":
                                float(
                                    values[
                                        "Distance_to_Road_m"
                                    ][idx]
                                ),
                        }


                        for k in range(
                            1,
                            12,
                        ):

                            row[
                                f"Lithology_Group_{k}"
                            ] = int(
                                lith == k
                            )


                        for code in [
                            10,
                            20,
                            30,
                            40,
                            50,
                            60,
                            80,
                            90,
                        ]:

                            row[
                                f"LULC_WorldCover_{code}"
                            ] = int(
                                lulc == code
                            )


                        sampled[
                            set_id
                        ].append(row)


                seen[s] = end


    # =========================================================================
    # VERIFY AND WRITE FINAL MATRICES
    # =========================================================================

    print()
    print("=" * 105)
    print("FINAL DATASET VALIDATION")
    print("=" * 105)


    balance_rows = []


    for set_id in range(
        1,
        6,
    ):

        bg = pd.DataFrame(
            sampled[
                set_id
            ]
        )


        if len(bg) != 3408:

            raise RuntimeError(
                f"Set {set_id}: expected 3,408 backgrounds, "
                f"found {len(bg)}"
            )


        if bg["Cell_ID"].duplicated().any():

            raise RuntimeError(
                f"Set {set_id}: duplicate background cells."
            )


        # -------------------------------------------------------------
        # Positive rows
        # -------------------------------------------------------------

        pos = positive.copy()

        pos[
            "Dataset_Set"
        ] = set_id

        pos[
            "Road_Match_Stratum"
        ] = road_stratum(
            pos[
                "Distance_to_Road_m"
            ].to_numpy()
        )


        # Keep only columns required by new matrix.
        meta_cols = [
            "Dataset_Set",
            "Cell_ID",
            "Raster_Row",
            "Raster_Col",
            "CV_Fold",
            "Label",
            "Road_Match_Stratum",
        ]


        final_columns = (
            meta_cols
            + FEATURES
        )


        # Ensure positive matrix has all needed columns.
        missing = (
            set(final_columns)
            - set(pos.columns)
        )


        if missing:

            raise RuntimeError(
                f"Positive matrix missing columns: "
                f"{sorted(missing)}"
            )


        bg_final = bg[
            final_columns
        ].copy()

        pos_final = pos[
            final_columns
        ].copy()


        final = pd.concat(
            [
                pos_final,
                bg_final,
            ],
            ignore_index=True,
        )


        # Deterministic shuffle.
        final = final.sample(
            frac=1,
            random_state=SEEDS[
                set_id - 1
            ],
        ).reset_index(drop=True)


        if len(final) != 4544:

            raise RuntimeError(
                f"Set {set_id}: final row count != 4544"
            )


        if int(
            final["Label"].sum()
        ) != 1136:

            raise RuntimeError(
                f"Set {set_id}: positive count mismatch"
            )


        # -------------------------------------------------------------
        # No block leakage
        # -------------------------------------------------------------

        folds_per_block = (
            final
            .groupby(
                [
                    "Raster_Row",
                    "Raster_Col",
                ]
            )["CV_Fold"]
            .nunique()
            .max()
        )


        # -------------------------------------------------------------
        # Road balance by stratum
        # -------------------------------------------------------------

        pos_counts = (
            final[
                final["Label"] == 1
            ][
                "Road_Match_Stratum"
            ]
            .value_counts()
            .reindex(
                STRATA_ORDER,
                fill_value=0,
            )
        )


        bg_counts = (
            final[
                final["Label"] == 0
            ][
                "Road_Match_Stratum"
            ]
            .value_counts()
            .reindex(
                STRATA_ORDER,
                fill_value=0,
            )
        )


        for s in STRATA_ORDER:

            if (
                bg_counts[s]
                != 3 * pos_counts[s]
            ):

                raise RuntimeError(
                    f"Set {set_id}: matching failed in {s}"
                )


        pos_road = final.loc[
            final["Label"] == 1,
            "Distance_to_Road_m",
        ]

        bg_road = final.loc[
            final["Label"] == 0,
            "Distance_to_Road_m",
        ]


        balance_rows.append({
            "Dataset_Set":
                set_id,

            "Rows":
                len(final),

            "Positive":
                int(
                    final["Label"].sum()
                ),

            "Background":
                int(
                    (final["Label"] == 0).sum()
                ),

            "Positive_Median_Road_m":
                float(
                    pos_road.median()
                ),

            "Background_Median_Road_m":
                float(
                    bg_road.median()
                ),

            "Positive_Exact0_pct":
                float(
                    100
                    * (
                        pos_road <= 1e-6
                    ).mean()
                ),

            "Background_Exact0_pct":
                float(
                    100
                    * (
                        bg_road <= 1e-6
                    ).mean()
                ),

            "Positive_Within50_pct":
                float(
                    100
                    * (
                        pos_road < 50
                    ).mean()
                ),

            "Background_Within50_pct":
                float(
                    100
                    * (
                        bg_road < 50
                    ).mean()
                ),
        })


        out_path = (
            OUT_DIR
            / (
                "manipur_baseline_v1_"
                f"road_matched_matrix_set_{set_id:02d}.csv"
            )
        )


        final.to_csv(
            out_path,
            index=False,
        )


        print()
        print(
            f"SET {set_id:02d} PASS ✅"
        )

        print(
            "Rows       :",
            f"{len(final):,}"
        )

        print(
            "Positive   :",
            f"{int((final['Label'] == 1).sum()):,}"
        )

        print(
            "Background :",
            f"{int((final['Label'] == 0).sum()):,}"
        )

        print(
            "CV folds   :",
            sorted(
                final[
                    "CV_Fold"
                ].unique()
            )
        )

        print(
            "Saved:",
            out_path
        )


    balance = pd.DataFrame(
        balance_rows
    )


    balance.to_csv(
        OUT_DIR
        / "road_matching_balance_summary.csv",
        index=False,
    )


    print()
    print("=" * 105)
    print("ROAD-DISTANCE BALANCE CHECK")
    print("=" * 105)

    print(
        balance.to_string(
            index=False
        )
    )


    print()
    print("=" * 105)
    print("ROAD-MATCHED DATASET GENERATION COMPLETE ✅")
    print("=" * 105)

    print(
        "Output directory:",
        OUT_DIR
    )


finally:

    mask_src.close()

    for src in sources.values():
        src.close()
```

```python
%%writefile scripts/91_modeling/run_road_matched_baseline_v1.py

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    log_loss,
    brier_score_loss,
    confusion_matrix,
)


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = (
    ROOT
    / "output"
    / "road_matched_model_matrices"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_baseline_v1_results"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",
    "Distance_to_Road_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


PARAMS = {
    "objective": "binary:logistic",

    "n_estimators": 300,
    "learning_rate": 0.05,
    "max_depth": 4,

    "min_child_weight": 1.0,
    "gamma": 0.0,

    "subsample": 0.8,
    "colsample_bytree": 0.8,

    "reg_alpha": 0.0,
    "reg_lambda": 1.0,

    "scale_pos_weight": 1.0,

    "tree_method": "hist",
    "device": "cpu",

    "eval_metric": "logloss",

    "random_state": 20260816,

    "n_jobs": -1,

    "verbosity": 0,
}


def calculate_metrics(
    y_true,
    probability,
):

    prediction = (
        probability >= 0.5
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1],
    ).ravel()

    specificity = (
        tn / (tn + fp)
    )

    return {
        "ROC_AUC":
            roc_auc_score(
                y_true,
                probability,
            ),

        "Average_Precision":
            average_precision_score(
                y_true,
                probability,
            ),

        "Accuracy_0_5":
            accuracy_score(
                y_true,
                prediction,
            ),

        "Balanced_Accuracy_0_5":
            balanced_accuracy_score(
                y_true,
                prediction,
            ),

        "Precision_0_5":
            precision_score(
                y_true,
                prediction,
                zero_division=0,
            ),

        "Recall_0_5":
            recall_score(
                y_true,
                prediction,
                zero_division=0,
            ),

        "Specificity_0_5":
            specificity,

        "F1_0_5":
            f1_score(
                y_true,
                prediction,
                zero_division=0,
            ),

        "MCC_0_5":
            matthews_corrcoef(
                y_true,
                prediction,
            ),

        "Log_Loss":
            log_loss(
                y_true,
                probability,
            ),

        "Brier_Score":
            brier_score_loss(
                y_true,
                probability,
            ),

        "TN":
            int(tn),

        "FP":
            int(fp),

        "FN":
            int(fn),

        "TP":
            int(tp),
    }


print("=" * 105)
print("MANIPUR XGBOOST — ROAD-MATCHED BACKGROUND SENSITIVITY EXPERIMENT")
print("=" * 105)

print("Datasets        : 5")
print("Spatial folds   : 5")
print("Total fits      : 25")
print("Features        :", len(FEATURES))
print("Road included   : YES")
print("Background      : ROAD-DISTANCE MATCHED")
print()


fold_rows = []
dataset_rows = []


for set_id in range(1, 6):

    matrix_path = (
        MATRIX_DIR
        / (
            "manipur_baseline_v1_"
            f"road_matched_matrix_set_{set_id:02d}.csv"
        )
    )


    print()
    print("-" * 105)
    print(
        f"DATASET SET {set_id:02d}"
    )
    print("-" * 105)


    df = pd.read_csv(
        matrix_path
    )


    # -----------------------------------------------------------------
    # Safety checks
    # -----------------------------------------------------------------

    if len(df) != 4544:

        raise RuntimeError(
            f"Set {set_id}: expected 4,544 rows, "
            f"found {len(df)}"
        )


    if int(
        (df["Label"] == 1).sum()
    ) != 1136:

        raise RuntimeError(
            f"Set {set_id}: positive count mismatch"
        )


    if int(
        (df["Label"] == 0).sum()
    ) != 3408:

        raise RuntimeError(
            f"Set {set_id}: background count mismatch"
        )


    missing = (
        set(FEATURES)
        - set(df.columns)
    )


    if missing:

        raise RuntimeError(
            f"Missing features: {sorted(missing)}"
        )


    if set(
        df["CV_Fold"].unique()
    ) != {1, 2, 3, 4, 5}:

        raise RuntimeError(
            f"Set {set_id}: unexpected CV folds"
        )


    oof_probability = np.full(
        len(df),
        np.nan,
        dtype=np.float64,
    )


    # -----------------------------------------------------------------
    # 5-fold spatial CV
    # -----------------------------------------------------------------

    for fold in range(1, 6):

        train_mask = (
            df["CV_Fold"] != fold
        )

        test_mask = (
            df["CV_Fold"] == fold
        )


        train_df = df.loc[
            train_mask
        ]

        test_df = df.loc[
            test_mask
        ]


        # -------------------------------------------------------------
        # Spatial leakage check
        # -------------------------------------------------------------

        train_blocks = set(
            train_df[
                [
                    "Raster_Row",
                    "Raster_Col",
                ]
            ]
            .itertuples(
                index=False,
                name=None,
            )
        )

        test_blocks = set(
            test_df[
                [
                    "Raster_Row",
                    "Raster_Col",
                ]
            ]
            .itertuples(
                index=False,
                name=None,
            )
        )


        # The CV fold itself is already block-based.
        # Verify no cell occurs in both train and test.
        overlap = (
            train_blocks
            & test_blocks
        )


        if overlap:

            raise RuntimeError(
                f"Set {set_id}, Fold {fold}: "
                "train/test cell overlap detected."
            )


        X_train = (
            train_df[
                FEATURES
            ]
            .astype(np.float32)
        )

        y_train = (
            train_df[
                "Label"
            ]
            .astype(int)
        )


        X_test = (
            test_df[
                FEATURES
            ]
            .astype(np.float32)
        )

        y_test = (
            test_df[
                "Label"
            ]
            .astype(int)
        )


        model = xgb.XGBClassifier(
            **PARAMS
        )


        model.fit(
            X_train,
            y_train,
        )


        probability = (
            model.predict_proba(
                X_test
            )[:, 1]
        )


        oof_probability[
            test_mask.to_numpy()
        ] = probability


        metrics = calculate_metrics(
            y_test,
            probability,
        )


        metrics.update({
            "Dataset_Set":
                set_id,

            "Fold":
                fold,

            "Train_Rows":
                len(train_df),

            "Test_Rows":
                len(test_df),
        })


        fold_rows.append(
            metrics
        )


        print(
            f"Fold {fold} | "
            f"ROC-AUC={metrics['ROC_AUC']:.6f} | "
            f"AP={metrics['Average_Precision']:.6f} | "
            f"F1={metrics['F1_0_5']:.6f} | "
            f"Recall={metrics['Recall_0_5']:.6f}"
        )


    # -----------------------------------------------------------------
    # Pooled out-of-fold metrics for this realization
    # -----------------------------------------------------------------

    if np.isnan(
        oof_probability
    ).any():

        raise RuntimeError(
            f"Set {set_id}: missing OOF predictions."
        )


    y_all = (
        df["Label"]
        .astype(int)
        .to_numpy()
    )


    dataset_metrics = calculate_metrics(
        y_all,
        oof_probability,
    )


    dataset_metrics[
        "Dataset_Set"
    ] = set_id


    dataset_rows.append(
        dataset_metrics
    )


    # Save OOF predictions.
    oof_df = df[
        [
            "Cell_ID",
            "Raster_Row",
            "Raster_Col",
            "CV_Fold",
            "Label",
            "Distance_to_Road_m",
            "Road_Match_Stratum",
        ]
    ].copy()


    oof_df[
        "OOF_Probability"
    ] = oof_probability


    oof_df.to_csv(
        OUT_DIR
        / f"road_matched_oof_predictions_set_{set_id:02d}.csv",
        index=False,
    )


    print()

    print(
        f"SET {set_id:02d} OOF | "
        f"ROC-AUC={dataset_metrics['ROC_AUC']:.6f} | "
        f"AP={dataset_metrics['Average_Precision']:.6f} | "
        f"F1={dataset_metrics['F1_0_5']:.6f} | "
        f"Recall={dataset_metrics['Recall_0_5']:.6f}"
    )


# =============================================================================
# SAVE RESULTS
# =============================================================================

fold_df = pd.DataFrame(
    fold_rows
)

dataset_df = pd.DataFrame(
    dataset_rows
)


fold_df.to_csv(
    OUT_DIR
    / "road_matched_fold_metrics.csv",
    index=False,
)


dataset_df.to_csv(
    OUT_DIR
    / "road_matched_dataset_oof_metrics.csv",
    index=False,
)


metrics_to_summarize = [
    "ROC_AUC",
    "Average_Precision",
    "Accuracy_0_5",
    "Balanced_Accuracy_0_5",
    "Precision_0_5",
    "Recall_0_5",
    "Specificity_0_5",
    "F1_0_5",
    "MCC_0_5",
    "Log_Loss",
    "Brier_Score",
]


summary_rows = []


for metric in metrics_to_summarize:

    values = dataset_df[
        metric
    ]

    summary_rows.append({
        "Metric":
            metric,

        "Mean":
            values.mean(),

        "SD":
            values.std(
                ddof=1
            ),

        "Min":
            values.min(),

        "Max":
            values.max(),
    })


summary_df = pd.DataFrame(
    summary_rows
)


summary_df.to_csv(
    OUT_DIR
    / "road_matched_primary_summary.csv",
    index=False,
)


print()
print("=" * 105)
print("ROAD-MATCHED MODEL — FINAL SUMMARY")
print("=" * 105)

print(
    summary_df.to_string(
        index=False
    )
)


# =============================================================================
# REFERENCE COMPARISON
# =============================================================================

original_reference = {
    "ROC_AUC":
        0.923455,

    "Average_Precision":
        0.861927,

    "Balanced_Accuracy_0_5":
        0.836913,

    "Precision_0_5":
        0.880890,

    "Recall_0_5":
        0.705634,

    "F1_0_5":
        0.783577,

    "MCC_0_5":
        0.729082,
}


no_road_reference = {
    "ROC_AUC":
        0.790653,

    "Average_Precision":
        0.560769,

    "Balanced_Accuracy_0_5":
        0.630604,

    "Precision_0_5":
        0.613419,

    "Recall_0_5":
        0.330634,

    "F1_0_5":
        0.429601,

    "MCC_0_5":
        0.331212,
}


print()
print("=" * 105)
print("THREE-WAY COMPARISON")
print("=" * 105)

print(
    f"{'Metric':28s}"
    f"{'Original':>14s}"
    f"{'No Road':>14s}"
    f"{'Road Matched':>16s}"
)


for metric in original_reference:

    road_matched_value = float(
        dataset_df[
            metric
        ].mean()
    )

    print(
        f"{metric:28s}"
        f"{original_reference[metric]:14.6f}"
        f"{no_road_reference[metric]:14.6f}"
        f"{road_matched_value:16.6f}"
    )


print()
print("=" * 105)
print("ROAD-MATCHED SENSITIVITY EXPERIMENT COMPLETE ✅")
print("=" * 105)

print(
    "Output:",
    OUT_DIR
)
```

```python
%%writefile scripts/91_modeling/train_road_matched_no_road_final_models.py

from pathlib import Path
import json

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = (
    ROOT
    / "output"
    / "road_matched_model_matrices"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_final_models"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ================================================================
# FROZEN 24-FEATURE ORDER
# Distance_to_Road_m deliberately excluded
# ================================================================

FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


PARAMS = {
    "objective": "binary:logistic",
    "n_estimators": 300,
    "learning_rate": 0.05,
    "max_depth": 4,
    "min_child_weight": 1.0,
    "gamma": 0.0,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "reg_alpha": 0.0,
    "reg_lambda": 1.0,
    "scale_pos_weight": 1.0,
    "tree_method": "hist",
    "device": "cpu",
    "eval_metric": "logloss",
    "random_state": 20260816,
    "n_jobs": -1,
    "verbosity": 0,
}


print("=" * 100)
print("ROAD-MATCHED + NO-ROAD — FINAL MODEL TRAINING")
print("=" * 100)

print("Models              : 5")
print("Features            :", len(FEATURES))
print("Distance to Road    : REMOVED")
print("Background sampling : ROAD-DISTANCE MATCHED")
print()


manifest_models = []


for set_id in range(1, 6):

    print("-" * 100)
    print(f"MODEL {set_id}/5")
    print("-" * 100)

    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_road_matched_matrix_set_{set_id:02d}.csv"
    )

    df = pd.read_csv(
        matrix_path
    )


    # ============================================================
    # VALIDATION
    # ============================================================

    if len(df) != 4544:
        raise RuntimeError(
            f"Set {set_id}: expected 4,544 rows, found {len(df)}"
        )


    positives = int(
        (df["Label"] == 1).sum()
    )

    backgrounds = int(
        (df["Label"] == 0).sum()
    )


    if positives != 1136:
        raise RuntimeError(
            f"Set {set_id}: expected 1,136 positives"
        )


    if backgrounds != 3408:
        raise RuntimeError(
            f"Set {set_id}: expected 3,408 backgrounds"
        )


    missing = (
        set(FEATURES)
        - set(df.columns)
    )


    if missing:
        raise RuntimeError(
            f"Missing features: {sorted(missing)}"
        )


    if "Distance_to_Road_m" in FEATURES:
        raise RuntimeError(
            "Distance_to_Road_m accidentally included."
        )


    X = (
        df[FEATURES]
        .astype(np.float32)
    )

    y = (
        df["Label"]
        .astype(int)
    )


    if X.isna().any().any():
        raise RuntimeError(
            f"Set {set_id}: NaN values detected."
        )


    # ============================================================
    # TRAIN FULL MODEL
    # ============================================================

    model = xgb.XGBClassifier(
        **PARAMS
    )

    model.fit(
        X,
        y,
    )


    # ============================================================
    # SAVE MODEL
    # ============================================================

    model_path = (
        OUT_DIR
        / f"manipur_xgb_road_matched_no_road_set_{set_id:02d}.json"
    )


    model.save_model(
        model_path
    )


    # ============================================================
    # VERIFY MODEL CAN BE RELOADED
    # ============================================================

    test_model = xgb.XGBClassifier()

    test_model.load_model(
        model_path
    )


    test_prob = (
        test_model.predict_proba(
            X.iloc[:10]
        )[:, 1]
    )


    if not np.all(
        np.isfinite(test_prob)
    ):
        raise RuntimeError(
            f"Set {set_id}: invalid predictions after reload."
        )


    manifest_models.append({
        "Dataset_Set":
            set_id,

        "Training_Rows":
            len(df),

        "Positive":
            positives,

        "Background":
            backgrounds,

        "Feature_Count":
            len(FEATURES),

        "Model_File":
            model_path.name,
    })


    print(
        "Rows       :",
        f"{len(df):,}"
    )

    print(
        "Positive   :",
        f"{positives:,}"
    )

    print(
        "Background :",
        f"{backgrounds:,}"
    )

    print(
        "Features   :",
        len(FEATURES)
    )

    print(
        "Saved      :",
        model_path
    )

    print(
        "Reload test: PASS ✅"
    )

    print()


# ================================================================
# SAVE FEATURE ORDER
# ================================================================

feature_path = (
    OUT_DIR
    / "road_matched_no_road_feature_order.json"
)

with open(
    feature_path,
    "w",
) as f:

    json.dump(
        FEATURES,
        f,
        indent=2,
    )


# ================================================================
# SAVE MANIFEST
# ================================================================

manifest = {
    "Model_Name":
        "Manipur Road-Matched No-Road XGBoost",

    "Purpose":
        "Bias-controlled landslide susceptibility sensitivity model",

    "Training_Datasets":
        5,

    "Rows_Per_Dataset":
        4544,

    "Positive_Per_Dataset":
        1136,

    "Background_Per_Dataset":
        3408,

    "Background_Strategy":
        "Road-distance matched",

    "Distance_to_Road_Feature":
        False,

    "Feature_Count":
        len(FEATURES),

    "Features":
        FEATURES,

    "XGBoost_Parameters":
        PARAMS,

    "Models":
        manifest_models,
}


manifest_path = (
    OUT_DIR
    / "road_matched_no_road_manifest.json"
)


with open(
    manifest_path,
    "w",
) as f:

    json.dump(
        manifest,
        f,
        indent=2,
    )


print("=" * 100)
print("FINAL MODEL TRAINING COMPLETE ✅")
print("=" * 100)

print()
print("Models saved to:")
print(OUT_DIR)

print()
print("Feature count:", len(FEATURES))

print()
print("Next step: SHAP interpretation of these 5 models.")
```

```python
%%writefile scripts/91_modeling/analyze_road_matched_no_road_shap.py

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = (
    ROOT
    / "output"
    / "road_matched_model_matrices"
)

MODEL_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_final_models"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_shap"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# =====================================================================
# EXACT 24-FEATURE ORDER
# =====================================================================

FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


FACTOR_GROUPS = {
    "Elevation": [
        "Elevation_m"
    ],

    "Slope": [
        "Slope_deg"
    ],

    "TWI": [
        "TWI"
    ],

    "Rainfall": [
        "Rainfall_CHELSA_Annual_mm"
    ],

    "Distance to River": [
        "Distance_to_River_m"
    ],

    "Lithology": [
        "Lithology_Group_1",
        "Lithology_Group_2",
        "Lithology_Group_3",
        "Lithology_Group_4",
        "Lithology_Group_5",
        "Lithology_Group_6",
        "Lithology_Group_7",
        "Lithology_Group_8",
        "Lithology_Group_9",
        "Lithology_Group_10",
        "Lithology_Group_11",
    ],

    "LULC": [
        "LULC_WorldCover_10",
        "LULC_WorldCover_20",
        "LULC_WorldCover_30",
        "LULC_WorldCover_40",
        "LULC_WorldCover_50",
        "LULC_WorldCover_60",
        "LULC_WorldCover_80",
        "LULC_WorldCover_90",
    ],
}


print("=" * 105)
print("ROAD-MATCHED + NO-ROAD — SHAP ANALYSIS")
print("=" * 105)

print("Models            : 5")
print("Features          :", len(FEATURES))
print("Physical factors  :", len(FACTOR_GROUPS))
print("Distance to Road  : REMOVED")
print()


feature_results = []
factor_results = []


for set_id in range(1, 6):

    print("-" * 105)
    print(f"MODEL {set_id}/5")
    print("-" * 105)


    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_road_matched_matrix_set_{set_id:02d}.csv"
    )

    model_path = (
        MODEL_DIR
        / f"manipur_xgb_road_matched_no_road_set_{set_id:02d}.json"
    )


    df = pd.read_csv(
        matrix_path
    )


    if len(df) != 4544:
        raise RuntimeError(
            f"Set {set_id}: expected 4,544 rows."
        )


    missing = (
        set(FEATURES)
        - set(df.columns)
    )

    if missing:
        raise RuntimeError(
            f"Set {set_id}: missing features {sorted(missing)}"
        )


    X = (
        df[FEATURES]
        .astype(np.float32)
    )


    # =================================================================
    # LOAD XGBOOST BOOSTER
    # =================================================================

    booster = xgb.Booster()

    booster.load_model(
        str(model_path)
    )


    dmatrix = xgb.DMatrix(
        X,
        feature_names=FEATURES,
    )


    # =================================================================
    # TREE SHAP
    # Last column = base value / expected value
    # =================================================================

    contributions = booster.predict(
        dmatrix,
        pred_contribs=True,
    )


    expected_columns = (
        len(FEATURES) + 1
    )


    if contributions.shape != (
        len(df),
        expected_columns,
    ):

        raise RuntimeError(
            f"Unexpected SHAP shape: {contributions.shape}"
        )


    shap_values = contributions[
        :,
        :-1
    ]


    base_values = contributions[
        :,
        -1
    ]


    # =================================================================
    # ADDITIVITY VALIDATION
    #
    # base + SHAP contributions should reconstruct raw model margin.
    # =================================================================

    raw_margin = booster.predict(
        dmatrix,
        output_margin=True,
    )


    reconstructed_margin = (
        base_values
        + shap_values.sum(axis=1)
    )


    max_additivity_error = float(
        np.max(
            np.abs(
                raw_margin
                - reconstructed_margin
            )
        )
    )


    print(
        "SHAP matrix shape      :",
        shap_values.shape
    )

    print(
        "Mean base value        :",
        f"{base_values.mean():.6f}"
    )

    print(
        "Max additivity error   :",
        f"{max_additivity_error:.10f}"
    )


    if max_additivity_error > 1e-4:

        raise RuntimeError(
            f"Set {set_id}: SHAP additivity check failed."
        )


    # =================================================================
    # FEATURE-LEVEL IMPORTANCE
    # =================================================================

    mean_abs = np.mean(
        np.abs(shap_values),
        axis=0,
    )


    mean_signed = np.mean(
        shap_values,
        axis=0,
    )


    feature_df = pd.DataFrame({
        "Dataset_Set":
            set_id,

        "Feature":
            FEATURES,

        "Mean_Abs_SHAP":
            mean_abs,

        "Mean_Signed_SHAP":
            mean_signed,
    })


    feature_df = (
        feature_df
        .sort_values(
            "Mean_Abs_SHAP",
            ascending=False,
        )
        .reset_index(drop=True)
    )


    feature_df[
        "Rank"
    ] = np.arange(
        1,
        len(feature_df) + 1,
    )


    feature_df.to_csv(
        OUT_DIR
        / f"shap_feature_importance_set_{set_id:02d}.csv",
        index=False,
    )


    feature_results.append(
        feature_df
    )


    print()
    print("Top 10 features:")

    for row in feature_df.head(10).itertuples():

        print(
            f"{row.Feature:32s} | "
            f"Mean |SHAP| = {row.Mean_Abs_SHAP:.6f}"
        )


    # =================================================================
    # PHYSICAL-FACTOR SHAP
    #
    # Two factor-level measures are saved:
    #
    # 1. Mean_Abs_Net_SHAP:
    #    Mean absolute value of the SUM of SHAP values belonging
    #    to the physical factor.
    #
    #    This is the main factor-level measure.
    #
    # 2. Sum_Mean_Abs_Feature_SHAP:
    #    Sum of individual feature mean |SHAP| values.
    #
    #    Useful as supplementary information for one-hot groups.
    # =================================================================

    feature_index = {
        name: i
        for i, name in enumerate(FEATURES)
    }


    factor_rows = []


    for factor, members in FACTOR_GROUPS.items():

        indexes = [
            feature_index[
                feature
            ]
            for feature in members
        ]


        group_values = shap_values[
            :,
            indexes
        ]


        # Net signed contribution of the whole factor for each row.
        net_group_shap = (
            group_values.sum(axis=1)
        )


        mean_abs_net = float(
            np.mean(
                np.abs(
                    net_group_shap
                )
            )
        )


        mean_signed_net = float(
            np.mean(
                net_group_shap
            )
        )


        sum_mean_abs = float(
            np.mean(
                np.abs(
                    group_values
                ),
                axis=0,
            ).sum()
        )


        factor_rows.append({
            "Dataset_Set":
                set_id,

            "Physical_Factor":
                factor,

            "Mean_Abs_Net_SHAP":
                mean_abs_net,

            "Mean_Signed_Net_SHAP":
                mean_signed_net,

            "Sum_Mean_Abs_Feature_SHAP":
                sum_mean_abs,

            "Encoded_Features":
                len(members),
        })


    factor_df = pd.DataFrame(
        factor_rows
    )


    factor_results.append(
        factor_df
    )


    print()


# =====================================================================
# GLOBAL 24-FEATURE IMPORTANCE
# =====================================================================

all_features = pd.concat(
    feature_results,
    ignore_index=True,
)


global_features = (
    all_features
    .groupby(
        "Feature",
        as_index=False,
    )
    .agg(
        Mean_Abs_SHAP_5_Models=(
            "Mean_Abs_SHAP",
            "mean",
        ),

        SD_Abs_SHAP_5_Models=(
            "Mean_Abs_SHAP",
            "std",
        ),

        Mean_Signed_SHAP_5_Models=(
            "Mean_Signed_SHAP",
            "mean",
        ),
    )
    .sort_values(
        "Mean_Abs_SHAP_5_Models",
        ascending=False,
    )
    .reset_index(drop=True)
)


global_features.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(global_features) + 1,
    ),
)


global_features.to_csv(
    OUT_DIR
    / "shap_global_importance_24_features.csv",
    index=False,
)


# =====================================================================
# GLOBAL 7-FACTOR IMPORTANCE
# =====================================================================

all_factors = pd.concat(
    factor_results,
    ignore_index=True,
)


global_factors = (
    all_factors
    .groupby(
        "Physical_Factor",
        as_index=False,
    )
    .agg(
        Mean_Abs_Net_SHAP_5_Models=(
            "Mean_Abs_Net_SHAP",
            "mean",
        ),

        SD_Abs_Net_SHAP_5_Models=(
            "Mean_Abs_Net_SHAP",
            "std",
        ),

        Mean_Signed_Net_SHAP_5_Models=(
            "Mean_Signed_Net_SHAP",
            "mean",
        ),

        Mean_Sum_Abs_Feature_SHAP=(
            "Sum_Mean_Abs_Feature_SHAP",
            "mean",
        ),

        Encoded_Features=(
            "Encoded_Features",
            "first",
        ),
    )
    .sort_values(
        "Mean_Abs_Net_SHAP_5_Models",
        ascending=False,
    )
    .reset_index(drop=True)
)


global_factors.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(global_factors) + 1,
    ),
)


global_factors.to_csv(
    OUT_DIR
    / "shap_global_importance_7_physical_factors.csv",
    index=False,
)


# =====================================================================
# PRINT FINAL RESULTS
# =====================================================================

print()
print("=" * 105)
print("GLOBAL SHAP IMPORTANCE — 24 FEATURES")
print("=" * 105)

print(
    global_features[
        [
            "Rank",
            "Feature",
            "Mean_Abs_SHAP_5_Models",
            "SD_Abs_SHAP_5_Models",
        ]
    ].to_string(
        index=False
    )
)


print()
print("=" * 105)
print("GLOBAL SHAP IMPORTANCE — 7 PHYSICAL FACTORS")
print("=" * 105)

print(
    global_factors[
        [
            "Rank",
            "Physical_Factor",
            "Mean_Abs_Net_SHAP_5_Models",
            "SD_Abs_Net_SHAP_5_Models",
            "Mean_Sum_Abs_Feature_SHAP",
        ]
    ].to_string(
        index=False
    )
)


print()
print("=" * 105)
print("SHAP ANALYSIS COMPLETE ✅")
print("=" * 105)

print()
print("Output directory:")
print(OUT_DIR)

print()
print(
    "IMPORTANT: SHAP values are on the XGBoost raw-margin/log-odds scale."
)

print(
    "Mean absolute SHAP measures model influence, not causality."
)

print(
    "Do not infer whether higher/lower feature values increase susceptibility "
    "from global importance alone."
)
```

```python
%%writefile scripts/91_modeling/shap_dependence_slope_bias_controlled.py

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = (
    ROOT
    / "output"
    / "road_matched_model_matrices"
)

MODEL_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_final_models"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_shap"
    / "dependence"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


FEATURE = "Slope_deg"
FEATURE_INDEX = FEATURES.index(FEATURE)


BINS = [
    0,
    5,
    10,
    15,
    20,
    25,
    30,
    35,
    40,
    45,
    50,
    60,
    np.inf,
]


LABELS = [
    "0-5",
    "5-10",
    "10-15",
    "15-20",
    "20-25",
    "25-30",
    "30-35",
    "35-40",
    "40-45",
    "45-50",
    "50-60",
    ">60",
]


print("=" * 100)
print("BIAS-CONTROLLED SHAP DEPENDENCE — SLOPE")
print("=" * 100)


all_values = []
model_summaries = []


for set_id in range(1, 6):

    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_road_matched_matrix_set_{set_id:02d}.csv"
    )

    model_path = (
        MODEL_DIR
        / f"manipur_xgb_road_matched_no_road_set_{set_id:02d}.json"
    )


    df = pd.read_csv(matrix_path)

    X = df[FEATURES].astype(
        np.float32
    )


    booster = xgb.Booster()

    booster.load_model(
        str(model_path)
    )


    dmatrix = xgb.DMatrix(
        X,
        feature_names=FEATURES,
    )


    contributions = booster.predict(
        dmatrix,
        pred_contribs=True,
    )


    slope_shap = contributions[
        :,
        FEATURE_INDEX
    ]


    temp = pd.DataFrame({
        "Dataset_Set":
            set_id,

        "Slope_deg":
            X["Slope_deg"].to_numpy(),

        "Slope_SHAP":
            slope_shap,

        "Label":
            df["Label"].astype(int).to_numpy(),
    })


    temp["Slope_Band"] = pd.cut(
        temp["Slope_deg"],
        bins=BINS,
        labels=LABELS,
        include_lowest=True,
        right=False,
    )


    all_values.append(temp)


    summary = (
        temp
        .groupby(
            "Slope_Band",
            observed=True,
        )
        .agg(
            Samples=(
                "Slope_deg",
                "size",
            ),

            Median_Slope_deg=(
                "Slope_deg",
                "median",
            ),

            Mean_SHAP=(
                "Slope_SHAP",
                "mean",
            ),

            Median_SHAP=(
                "Slope_SHAP",
                "median",
            ),

            Mean_Abs_SHAP=(
                "Slope_SHAP",
                lambda x: np.mean(np.abs(x)),
            ),
        )
        .reset_index()
    )


    summary[
        "Dataset_Set"
    ] = set_id


    model_summaries.append(summary)


combined = pd.concat(
    all_values,
    ignore_index=True,
)


combined.to_csv(
    OUT_DIR
    / "slope_shap_values.csv",
    index=False,
)


per_model = pd.concat(
    model_summaries,
    ignore_index=True,
)


per_model.to_csv(
    OUT_DIR
    / "slope_shap_summary_per_model.csv",
    index=False,
)


# ---------------------------------------------------------------------
# Aggregate across five models
# ---------------------------------------------------------------------

global_summary = (
    per_model
    .groupby(
        "Slope_Band",
        observed=True,
    )
    .agg(
        Mean_Samples=(
            "Samples",
            "mean",
        ),

        Median_Slope_deg=(
            "Median_Slope_deg",
            "mean",
        ),

        Mean_SHAP_5_Models=(
            "Mean_SHAP",
            "mean",
        ),

        SD_Mean_SHAP_5_Models=(
            "Mean_SHAP",
            "std",
        ),

        Median_SHAP_5_Models=(
            "Median_SHAP",
            "mean",
        ),

        Mean_Abs_SHAP_5_Models=(
            "Mean_Abs_SHAP",
            "mean",
        ),
    )
    .reset_index()
)


global_summary.to_csv(
    OUT_DIR
    / "slope_shap_dependence_summary.csv",
    index=False,
)


# ---------------------------------------------------------------------
# Correlations
# ---------------------------------------------------------------------

spearman = (
    combined[
        [
            "Slope_deg",
            "Slope_SHAP",
        ]
    ]
    .corr(
        method="spearman"
    )
    .iloc[0, 1]
)


pearson = (
    combined[
        [
            "Slope_deg",
            "Slope_SHAP",
        ]
    ]
    .corr(
        method="pearson"
    )
    .iloc[0, 1]
)


print()
print("=" * 100)
print("SLOPE SHAP DEPENDENCE SUMMARY")
print("=" * 100)

print(
    global_summary.to_string(
        index=False
    )
)


print()
print(
    "Spearman correlation "
    "(Slope vs SHAP):",
    f"{spearman:.4f}",
)

print(
    "Pearson correlation  "
    "(Slope vs SHAP):",
    f"{pearson:.4f}",
)


print()
print("Interpretation:")

print(
    "SHAP > 0 = slope range pushes prediction "
    "toward HIGHER susceptibility"
)

print(
    "SHAP < 0 = slope range pushes prediction "
    "toward LOWER susceptibility"
)

print(
    "SHAP values are contributions on the "
    "XGBoost raw-margin/log-odds scale."
)


print()
print("=" * 100)
print("SLOPE DEPENDENCE ANALYSIS COMPLETE ✅")
print("=" * 100)
```

```python
%%writefile scripts/91_modeling/shap_dependence_twi_bias_controlled.py

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = (
    ROOT
    / "output"
    / "road_matched_model_matrices"
)

MODEL_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_final_models"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_shap"
    / "dependence"
)

OUT_DIR.mkdir(parents=True, exist_ok=True)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


FEATURE = "TWI"
FEATURE_INDEX = FEATURES.index(FEATURE)


BINS = [
    -np.inf,
    4.5,
    5,
    6,
    7,
    8,
    10,
    12,
    14,
    16,
    np.inf,
]


LABELS = [
    "<4.5",
    "4.5-5",
    "5-6",
    "6-7",
    "7-8",
    "8-10",
    "10-12",
    "12-14",
    "14-16",
    ">16",
]


print("=" * 100)
print("BIAS-CONTROLLED SHAP DEPENDENCE — TWI")
print("=" * 100)


all_values = []
model_summaries = []


for set_id in range(1, 6):

    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_road_matched_matrix_set_{set_id:02d}.csv"
    )

    model_path = (
        MODEL_DIR
        / f"manipur_xgb_road_matched_no_road_set_{set_id:02d}.json"
    )


    df = pd.read_csv(matrix_path)

    X = df[FEATURES].astype(np.float32)


    booster = xgb.Booster()
    booster.load_model(str(model_path))


    dmatrix = xgb.DMatrix(
        X,
        feature_names=FEATURES,
    )


    contributions = booster.predict(
        dmatrix,
        pred_contribs=True,
    )


    twi_shap = contributions[
        :,
        FEATURE_INDEX
    ]


    temp = pd.DataFrame({
        "Dataset_Set": set_id,
        "TWI": X["TWI"].to_numpy(),
        "TWI_SHAP": twi_shap,
        "Label": df["Label"].astype(int).to_numpy(),
    })


    temp["TWI_Band"] = pd.cut(
        temp["TWI"],
        bins=BINS,
        labels=LABELS,
        include_lowest=True,
        right=False,
    )


    all_values.append(temp)


    summary = (
        temp
        .groupby(
            "TWI_Band",
            observed=True,
        )
        .agg(
            Samples=(
                "TWI",
                "size",
            ),

            Median_TWI=(
                "TWI",
                "median",
            ),

            Mean_SHAP=(
                "TWI_SHAP",
                "mean",
            ),

            Median_SHAP=(
                "TWI_SHAP",
                "median",
            ),

            Mean_Abs_SHAP=(
                "TWI_SHAP",
                lambda x: np.mean(np.abs(x)),
            ),
        )
        .reset_index()
    )


    summary["Dataset_Set"] = set_id

    model_summaries.append(summary)


combined = pd.concat(
    all_values,
    ignore_index=True,
)


combined.to_csv(
    OUT_DIR / "twi_shap_values.csv",
    index=False,
)


per_model = pd.concat(
    model_summaries,
    ignore_index=True,
)


per_model.to_csv(
    OUT_DIR / "twi_shap_summary_per_model.csv",
    index=False,
)


global_summary = (
    per_model
    .groupby(
        "TWI_Band",
        observed=True,
    )
    .agg(
        Mean_Samples=(
            "Samples",
            "mean",
        ),

        Median_TWI=(
            "Median_TWI",
            "mean",
        ),

        Mean_SHAP_5_Models=(
            "Mean_SHAP",
            "mean",
        ),

        SD_Mean_SHAP_5_Models=(
            "Mean_SHAP",
            "std",
        ),

        Median_SHAP_5_Models=(
            "Median_SHAP",
            "mean",
        ),

        Mean_Abs_SHAP_5_Models=(
            "Mean_Abs_SHAP",
            "mean",
        ),
    )
    .reset_index()
)


global_summary.to_csv(
    OUT_DIR / "twi_shap_dependence_summary.csv",
    index=False,
)


spearman = (
    combined[
        ["TWI", "TWI_SHAP"]
    ]
    .corr(method="spearman")
    .iloc[0, 1]
)


pearson = (
    combined[
        ["TWI", "TWI_SHAP"]
    ]
    .corr(method="pearson")
    .iloc[0, 1]
)


print()
print("=" * 100)
print("TWI SHAP DEPENDENCE SUMMARY")
print("=" * 100)

print(
    global_summary.to_string(
        index=False
    )
)


print()

print(
    "Spearman correlation (TWI vs SHAP):",
    f"{spearman:.4f}",
)

print(
    "Pearson correlation  (TWI vs SHAP):",
    f"{pearson:.4f}",
)


print()
print("Interpretation:")

print(
    "SHAP > 0 = TWI range pushes prediction "
    "toward HIGHER susceptibility"
)

print(
    "SHAP < 0 = TWI range pushes prediction "
    "toward LOWER susceptibility"
)

print(
    "SHAP values are on the XGBoost "
    "raw-margin/log-odds scale."
)


print()
print("=" * 100)
print("TWI DEPENDENCE ANALYSIS COMPLETE ✅")
print("=" * 100)
```

```python
%%writefile scripts/91_modeling/shap_dependence_river_bias_controlled.py

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = (
    ROOT
    / "output"
    / "road_matched_model_matrices"
)

MODEL_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_final_models"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_shap"
    / "dependence"
)

OUT_DIR.mkdir(parents=True, exist_ok=True)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


FEATURE = "Distance_to_River_m"
FEATURE_INDEX = FEATURES.index(FEATURE)


def river_band(values):

    values = np.asarray(values)

    result = np.empty(
        values.shape,
        dtype=object,
    )

    result[
        values <= 1e-6
    ] = "Exactly_0m"

    result[
        (values > 1e-6)
        & (values < 100)
    ] = "GT0_100m"

    result[
        (values >= 100)
        & (values < 250)
    ] = "100_250m"

    result[
        (values >= 250)
        & (values < 500)
    ] = "250_500m"

    result[
        (values >= 500)
        & (values < 750)
    ] = "500_750m"

    result[
        (values >= 750)
        & (values < 1000)
    ] = "750_1000m"

    result[
        (values >= 1000)
        & (values < 1250)
    ] = "1000_1250m"

    result[
        (values >= 1250)
        & (values < 1500)
    ] = "1250_1500m"

    result[
        (values >= 1500)
        & (values < 1750)
    ] = "1500_1750m"

    result[
        (values >= 1750)
        & (values < 2000)
    ] = "1750_2000m"

    result[
        (values >= 2000)
        & (values < 2250)
    ] = "2000_2250m"

    result[
        values >= 2250
    ] = "GE2250m"

    return result


BAND_ORDER = [
    "Exactly_0m",
    "GT0_100m",
    "100_250m",
    "250_500m",
    "500_750m",
    "750_1000m",
    "1000_1250m",
    "1250_1500m",
    "1500_1750m",
    "1750_2000m",
    "2000_2250m",
    "GE2250m",
]


print("=" * 105)
print("BIAS-CONTROLLED SHAP DEPENDENCE — DISTANCE TO RIVER")
print("=" * 105)


all_values = []
model_summaries = []


for set_id in range(1, 6):

    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_road_matched_matrix_set_{set_id:02d}.csv"
    )

    model_path = (
        MODEL_DIR
        / f"manipur_xgb_road_matched_no_road_set_{set_id:02d}.json"
    )


    df = pd.read_csv(matrix_path)

    X = df[FEATURES].astype(np.float32)


    booster = xgb.Booster()
    booster.load_model(str(model_path))


    dmatrix = xgb.DMatrix(
        X,
        feature_names=FEATURES,
    )


    contributions = booster.predict(
        dmatrix,
        pred_contribs=True,
    )


    river_shap = contributions[
        :,
        FEATURE_INDEX
    ]


    temp = pd.DataFrame({
        "Dataset_Set":
            set_id,

        "Distance_to_River_m":
            X["Distance_to_River_m"].to_numpy(),

        "River_SHAP":
            river_shap,

        "Label":
            df["Label"].astype(int).to_numpy(),
    })


    temp["River_Band"] = river_band(
        temp["Distance_to_River_m"].to_numpy()
    )


    temp["River_Band"] = pd.Categorical(
        temp["River_Band"],
        categories=BAND_ORDER,
        ordered=True,
    )


    all_values.append(temp)


    summary = (
        temp
        .groupby(
            "River_Band",
            observed=True,
        )
        .agg(
            Samples=(
                "Distance_to_River_m",
                "size",
            ),

            Median_Distance_m=(
                "Distance_to_River_m",
                "median",
            ),

            Mean_SHAP=(
                "River_SHAP",
                "mean",
            ),

            Median_SHAP=(
                "River_SHAP",
                "median",
            ),

            Mean_Abs_SHAP=(
                "River_SHAP",
                lambda x: np.mean(np.abs(x)),
            ),
        )
        .reset_index()
    )


    summary["Dataset_Set"] = set_id

    model_summaries.append(summary)


combined = pd.concat(
    all_values,
    ignore_index=True,
)


combined.to_csv(
    OUT_DIR / "river_shap_values.csv",
    index=False,
)


per_model = pd.concat(
    model_summaries,
    ignore_index=True,
)


per_model.to_csv(
    OUT_DIR / "river_shap_summary_per_model.csv",
    index=False,
)


# =====================================================================
# FIVE-MODEL SUMMARY
# =====================================================================

global_summary = (
    per_model
    .groupby(
        "River_Band",
        observed=True,
    )
    .agg(
        Mean_Samples=(
            "Samples",
            "mean",
        ),

        Median_Distance_m=(
            "Median_Distance_m",
            "mean",
        ),

        Mean_SHAP_5_Models=(
            "Mean_SHAP",
            "mean",
        ),

        SD_Mean_SHAP_5_Models=(
            "Mean_SHAP",
            "std",
        ),

        Median_SHAP_5_Models=(
            "Median_SHAP",
            "mean",
        ),

        Mean_Abs_SHAP_5_Models=(
            "Mean_Abs_SHAP",
            "mean",
        ),
    )
    .reset_index()
)


global_summary.to_csv(
    OUT_DIR
    / "river_shap_dependence_summary.csv",
    index=False,
)


# =====================================================================
# CORRELATIONS
# =====================================================================

spearman = (
    combined[
        [
            "Distance_to_River_m",
            "River_SHAP",
        ]
    ]
    .corr(method="spearman")
    .iloc[0, 1]
)


pearson = (
    combined[
        [
            "Distance_to_River_m",
            "River_SHAP",
        ]
    ]
    .corr(method="pearson")
    .iloc[0, 1]
)


# =====================================================================
# PRINT
# =====================================================================

print()
print("=" * 105)
print("DISTANCE-TO-RIVER SHAP DEPENDENCE SUMMARY")
print("=" * 105)


print(
    global_summary.to_string(
        index=False
    )
)


print()

print(
    "Spearman correlation "
    "(River Distance vs SHAP):",
    f"{spearman:.4f}",
)

print(
    "Pearson correlation  "
    "(River Distance vs SHAP):",
    f"{pearson:.4f}",
)


print()
print("Interpretation:")

print(
    "SHAP > 0 = river-distance range pushes prediction "
    "toward HIGHER susceptibility"
)

print(
    "SHAP < 0 = river-distance range pushes prediction "
    "toward LOWER susceptibility"
)

print(
    "SHAP values are on the XGBoost "
    "raw-margin/log-odds scale."
)


print()
print("=" * 105)
print("DISTANCE-TO-RIVER DEPENDENCE ANALYSIS COMPLETE ✅")
print("=" * 105)
```

```python
%%writefile scripts/91_modeling/shap_lithology_categories_bias_controlled.py

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = (
    ROOT
    / "output"
    / "road_matched_model_matrices"
)

MODEL_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_final_models"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_shap"
    / "categorical"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


LITHOLOGY_FEATURES = [
    f"Lithology_Group_{i}"
    for i in range(1, 12)
]


LITHOLOGY_INDEXES = [
    FEATURES.index(feature)
    for feature in LITHOLOGY_FEATURES
]


print("=" * 105)
print("BIAS-CONTROLLED SHAP — LITHOLOGY CATEGORY ANALYSIS")
print("=" * 105)


all_rows = []
per_model_summaries = []


for set_id in range(1, 6):

    print(
        f"Processing model {set_id}/5..."
    )

    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_road_matched_matrix_set_{set_id:02d}.csv"
    )

    model_path = (
        MODEL_DIR
        / f"manipur_xgb_road_matched_no_road_set_{set_id:02d}.json"
    )


    df = pd.read_csv(
        matrix_path
    )

    X = (
        df[FEATURES]
        .astype(np.float32)
    )


    # ================================================================
    # VALIDATE ONE-HOT LITHOLOGY
    # ================================================================

    lith_matrix = (
        X[LITHOLOGY_FEATURES]
        .to_numpy()
    )


    row_sums = lith_matrix.sum(
        axis=1
    )


    if not np.allclose(
        row_sums,
        1.0,
    ):

        bad = int(
            np.sum(
                ~np.isclose(
                    row_sums,
                    1.0,
                )
            )
        )

        raise RuntimeError(
            f"Set {set_id}: "
            f"{bad} rows do not have exactly one "
            "active lithology category."
        )


    active_index = np.argmax(
        lith_matrix,
        axis=1,
    )


    active_group = (
        active_index + 1
    )


    # ================================================================
    # TREE SHAP
    # ================================================================

    booster = xgb.Booster()

    booster.load_model(
        str(model_path)
    )


    dmatrix = xgb.DMatrix(
        X,
        feature_names=FEATURES,
    )


    contributions = booster.predict(
        dmatrix,
        pred_contribs=True,
    )


    shap_values = contributions[
        :,
        :-1
    ]


    lithology_shap_matrix = (
        shap_values[
            :,
            LITHOLOGY_INDEXES
        ]
    )


    # Net contribution of the entire lithology factor.
    net_lithology_shap = (
        lithology_shap_matrix.sum(
            axis=1
        )
    )


    # SHAP contribution of the active one-hot feature only.
    active_dummy_shap = (
        lithology_shap_matrix[
            np.arange(
                len(df)
            ),
            active_index,
        ]
    )


    temp = pd.DataFrame({
        "Dataset_Set":
            set_id,

        "Lithology_Group":
            active_group,

        "Net_Lithology_SHAP":
            net_lithology_shap,

        "Active_Dummy_SHAP":
            active_dummy_shap,

        "Label":
            df["Label"]
            .astype(int)
            .to_numpy(),
    })


    all_rows.append(
        temp
    )


    summary = (
        temp
        .groupby(
            "Lithology_Group",
            as_index=False,
        )
        .agg(
            Samples=(
                "Lithology_Group",
                "size",
            ),

            Mean_Net_SHAP=(
                "Net_Lithology_SHAP",
                "mean",
            ),

            Median_Net_SHAP=(
                "Net_Lithology_SHAP",
                "median",
            ),

            Mean_Abs_Net_SHAP=(
                "Net_Lithology_SHAP",
                lambda x:
                    np.mean(
                        np.abs(x)
                    ),
            ),

            Mean_Active_Dummy_SHAP=(
                "Active_Dummy_SHAP",
                "mean",
            ),

            Landslide_Rate=(
                "Label",
                "mean",
            ),
        )
    )


    summary[
        "Dataset_Set"
    ] = set_id


    per_model_summaries.append(
        summary
    )


# =====================================================================
# SAVE RAW VALUES
# =====================================================================

combined = pd.concat(
    all_rows,
    ignore_index=True,
)


combined.to_csv(
    OUT_DIR
    / "lithology_category_shap_values.csv",
    index=False,
)


per_model = pd.concat(
    per_model_summaries,
    ignore_index=True,
)


per_model.to_csv(
    OUT_DIR
    / "lithology_category_shap_per_model.csv",
    index=False,
)


# =====================================================================
# FIVE-MODEL AGGREGATION
# =====================================================================

global_summary = (
    per_model
    .groupby(
        "Lithology_Group",
        as_index=False,
    )
    .agg(
        Mean_Samples=(
            "Samples",
            "mean",
        ),

        Mean_Net_SHAP_5_Models=(
            "Mean_Net_SHAP",
            "mean",
        ),

        SD_Net_SHAP_5_Models=(
            "Mean_Net_SHAP",
            "std",
        ),

        Median_Net_SHAP_5_Models=(
            "Median_Net_SHAP",
            "mean",
        ),

        Mean_Abs_Net_SHAP_5_Models=(
            "Mean_Abs_Net_SHAP",
            "mean",
        ),

        Mean_Active_Dummy_SHAP=(
            "Mean_Active_Dummy_SHAP",
            "mean",
        ),

        Mean_Landslide_Rate=(
            "Landslide_Rate",
            "mean",
        ),
    )
)


global_summary[
    "Susceptibility_Direction"
] = np.where(
    global_summary[
        "Mean_Net_SHAP_5_Models"
    ] > 0,
    "Higher",
    np.where(
        global_summary[
            "Mean_Net_SHAP_5_Models"
        ] < 0,
        "Lower",
        "Neutral",
    ),
)


global_summary = (
    global_summary
    .sort_values(
        "Mean_Net_SHAP_5_Models",
        ascending=False,
    )
    .reset_index(drop=True)
)


global_summary.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(global_summary) + 1,
    ),
)


global_summary.to_csv(
    OUT_DIR
    / "lithology_category_shap_global_summary.csv",
    index=False,
)


print()
print("=" * 105)
print("LITHOLOGY CATEGORY SHAP SUMMARY")
print("=" * 105)


print(
    global_summary[
        [
            "Rank",
            "Lithology_Group",
            "Mean_Samples",
            "Mean_Net_SHAP_5_Models",
            "SD_Net_SHAP_5_Models",
            "Median_Net_SHAP_5_Models",
            "Mean_Abs_Net_SHAP_5_Models",
            "Susceptibility_Direction",
        ]
    ].to_string(
        index=False
    )
)


print()
print(
    "Positive net SHAP = lithology factor pushes "
    "prediction toward HIGHER susceptibility."
)

print(
    "Negative net SHAP = lithology factor pushes "
    "prediction toward LOWER susceptibility."
)

print(
    "Groups with very small sample counts must "
    "be interpreted cautiously."
)


print()
print("=" * 105)
print("LITHOLOGY CATEGORY ANALYSIS COMPLETE ✅")
print("=" * 105)
```

```python
%%writefile scripts/91_modeling/shap_lulc_categories_bias_controlled.py

from pathlib import Path

import numpy as np
import pandas as pd
import xgboost as xgb


ROOT = Path("/content/Manipur_XGBoost_Colab")

MATRIX_DIR = ROOT / "output" / "road_matched_model_matrices"

MODEL_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_final_models"
)

OUT_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_shap"
    / "categorical"
)

OUT_DIR.mkdir(parents=True, exist_ok=True)


FEATURES = [
    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


LULC_CODES = [
    10,
    20,
    30,
    40,
    50,
    60,
    80,
    90,
]


LULC_FEATURES = [
    f"LULC_WorldCover_{code}"
    for code in LULC_CODES
]


LULC_INDEXES = [
    FEATURES.index(feature)
    for feature in LULC_FEATURES
]


print("=" * 105)
print("BIAS-CONTROLLED SHAP — LULC CATEGORY ANALYSIS")
print("=" * 105)


all_rows = []
per_model_summaries = []


for set_id in range(1, 6):

    print(f"Processing model {set_id}/5...")

    matrix_path = (
        MATRIX_DIR
        / f"manipur_baseline_v1_road_matched_matrix_set_{set_id:02d}.csv"
    )

    model_path = (
        MODEL_DIR
        / f"manipur_xgb_road_matched_no_road_set_{set_id:02d}.json"
    )


    df = pd.read_csv(matrix_path)

    X = df[FEATURES].astype(np.float32)


    # ---------------------------------------------------------------
    # Validate one-hot LULC
    # ---------------------------------------------------------------

    lulc_matrix = (
        X[LULC_FEATURES]
        .to_numpy()
    )

    row_sums = lulc_matrix.sum(axis=1)


    if not np.allclose(row_sums, 1.0):

        bad = int(
            np.sum(
                ~np.isclose(
                    row_sums,
                    1.0,
                )
            )
        )

        raise RuntimeError(
            f"Set {set_id}: "
            f"{bad} rows do not have exactly one "
            "active LULC class."
        )


    active_index = np.argmax(
        lulc_matrix,
        axis=1,
    )


    active_code = np.array(
        LULC_CODES
    )[active_index]


    # ---------------------------------------------------------------
    # TreeSHAP
    # ---------------------------------------------------------------

    booster = xgb.Booster()

    booster.load_model(
        str(model_path)
    )


    dmatrix = xgb.DMatrix(
        X,
        feature_names=FEATURES,
    )


    contributions = booster.predict(
        dmatrix,
        pred_contribs=True,
    )


    shap_values = contributions[:, :-1]


    lulc_shap_matrix = (
        shap_values[:, LULC_INDEXES]
    )


    # Net SHAP contribution of entire LULC factor
    net_lulc_shap = (
        lulc_shap_matrix.sum(axis=1)
    )


    active_dummy_shap = (
        lulc_shap_matrix[
            np.arange(len(df)),
            active_index,
        ]
    )


    temp = pd.DataFrame({
        "Dataset_Set":
            set_id,

        "LULC_Code":
            active_code,

        "Net_LULC_SHAP":
            net_lulc_shap,

        "Active_Dummy_SHAP":
            active_dummy_shap,

        "Label":
            df["Label"]
            .astype(int)
            .to_numpy(),
    })


    all_rows.append(temp)


    summary = (
        temp
        .groupby(
            "LULC_Code",
            as_index=False,
        )
        .agg(
            Samples=(
                "LULC_Code",
                "size",
            ),

            Mean_Net_SHAP=(
                "Net_LULC_SHAP",
                "mean",
            ),

            Median_Net_SHAP=(
                "Net_LULC_SHAP",
                "median",
            ),

            Mean_Abs_Net_SHAP=(
                "Net_LULC_SHAP",
                lambda x:
                    np.mean(np.abs(x)),
            ),

            Mean_Active_Dummy_SHAP=(
                "Active_Dummy_SHAP",
                "mean",
            ),

            Landslide_Rate=(
                "Label",
                "mean",
            ),
        )
    )


    summary["Dataset_Set"] = set_id

    per_model_summaries.append(summary)


combined = pd.concat(
    all_rows,
    ignore_index=True,
)


combined.to_csv(
    OUT_DIR
    / "lulc_category_shap_values.csv",
    index=False,
)


per_model = pd.concat(
    per_model_summaries,
    ignore_index=True,
)


per_model.to_csv(
    OUT_DIR
    / "lulc_category_shap_per_model.csv",
    index=False,
)


global_summary = (
    per_model
    .groupby(
        "LULC_Code",
        as_index=False,
    )
    .agg(
        Mean_Samples=(
            "Samples",
            "mean",
        ),

        Mean_Net_SHAP_5_Models=(
            "Mean_Net_SHAP",
            "mean",
        ),

        SD_Net_SHAP_5_Models=(
            "Mean_Net_SHAP",
            "std",
        ),

        Median_Net_SHAP_5_Models=(
            "Median_Net_SHAP",
            "mean",
        ),

        Mean_Abs_Net_SHAP_5_Models=(
            "Mean_Abs_Net_SHAP",
            "mean",
        ),

        Mean_Active_Dummy_SHAP=(
            "Mean_Active_Dummy_SHAP",
            "mean",
        ),
    )
)


global_summary[
    "Susceptibility_Direction"
] = np.where(
    global_summary[
        "Mean_Net_SHAP_5_Models"
    ] > 0,
    "Higher",
    np.where(
        global_summary[
            "Mean_Net_SHAP_5_Models"
        ] < 0,
        "Lower",
        "Neutral",
    ),
)


global_summary = (
    global_summary
    .sort_values(
        "Mean_Net_SHAP_5_Models",
        ascending=False,
    )
    .reset_index(drop=True)
)


global_summary.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(global_summary) + 1,
    ),
)


global_summary.to_csv(
    OUT_DIR
    / "lulc_category_shap_global_summary.csv",
    index=False,
)


print()
print("=" * 105)
print("LULC CATEGORY SHAP SUMMARY")
print("=" * 105)


print(
    global_summary[
        [
            "Rank",
            "LULC_Code",
            "Mean_Samples",
            "Mean_Net_SHAP_5_Models",
            "SD_Net_SHAP_5_Models",
            "Median_Net_SHAP_5_Models",
            "Mean_Abs_Net_SHAP_5_Models",
            "Susceptibility_Direction",
        ]
    ].to_string(index=False)
)


print()
print(
    "Positive net SHAP = LULC factor pushes "
    "prediction toward HIGHER susceptibility."
)

print(
    "Negative net SHAP = LULC factor pushes "
    "prediction toward LOWER susceptibility."
)

print(
    "Classes with very small sample counts "
    "must be interpreted cautiously."
)


print()
print("=" * 105)
print("LULC CATEGORY ANALYSIS COMPLETE ✅")
print("=" * 105)
```

```python
%%writefile scripts/91_modeling/create_modelA_no_road_valid_mask.py

from pathlib import Path
import numpy as np
import rasterio
from rasterio.windows import Window


RASTER_DIR = Path(
    "/content/Manipur_XGBoost_Statewide_Rasters"
)

OUTPUT_DIR = (
    Path("/content/Manipur_XGBoost_Colab")
    / "output"
    / "statewide_prediction_modelA_bias_controlled"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


RASTERS = [
    RASTER_DIR / "manipur_dem_30m_utm46n.tif",
    RASTER_DIR / "manipur_slope_30m.tif",
    RASTER_DIR / "manipur_twi_30m.tif",
    RASTER_DIR / "manipur_rainfall_annual_30m.tif",
    RASTER_DIR / "manipur_distance_to_river_30m.tif",
    RASTER_DIR / "manipur_lithology_grouped_30m.tif",
    RASTER_DIR / "manipur_lulc_worldcover_2021_30m.tif",
]

OLD_MASK = (
    RASTER_DIR
    / "manipur_baseline_common_valid_mask_30m.tif"
)

NEW_MASK = (
    OUTPUT_DIR
    / "manipur_modelA_no_road_common_valid_mask_30m.tif"
)


print("=" * 90)
print("MODEL A — CREATE 7-FACTOR NO-ROAD VALID MASK")
print("=" * 90)


# ---------------------------------------------------------------------
# Open predictor rasters
# ---------------------------------------------------------------------

datasets = [
    rasterio.open(path)
    for path in RASTERS
]


reference = datasets[0]

width = reference.width
height = reference.height
crs = reference.crs
transform = reference.transform


# ---------------------------------------------------------------------
# Alignment validation
# ---------------------------------------------------------------------

for path, ds in zip(RASTERS, datasets):

    if ds.width != width or ds.height != height:
        raise RuntimeError(
            f"Shape mismatch: {path.name}"
        )

    if ds.crs != crs:
        raise RuntimeError(
            f"CRS mismatch: {path.name}"
        )

    if ds.transform != transform:
        raise RuntimeError(
            f"Transform mismatch: {path.name}"
        )


print("Alignment check: PASS ✅")
print("Shape:", (height, width))
print("CRS:", crs)
print("Transform:", transform)


# ---------------------------------------------------------------------
# Output profile
# ---------------------------------------------------------------------

profile = reference.profile.copy()

profile.update(
    dtype="uint8",
    count=1,
    nodata=0,
    compress="deflate",
    predictor=2,
    tiled=True,
    BIGTIFF="IF_SAFER",
)


new_valid_count = 0


# ---------------------------------------------------------------------
# Build mask block-by-block
# ---------------------------------------------------------------------

with rasterio.open(
    NEW_MASK,
    "w",
    **profile,
) as dst:

    for _, window in reference.block_windows(1):

        valid = np.ones(
            (
                int(window.height),
                int(window.width),
            ),
            dtype=bool,
        )

        for ds in datasets:

            data = ds.read(
                1,
                window=window,
                masked=True,
            )

            # Valid only when:
            # 1. not raster nodata/masked
            # 2. finite numeric value
            current_valid = (
                ~np.ma.getmaskarray(data)
                & np.isfinite(
                    np.asarray(
                        data.filled(np.nan),
                        dtype=np.float64,
                    )
                )
            )

            valid &= current_valid


        # -------------------------------------------------------------
        # Categorical sanity checks
        # -------------------------------------------------------------

        lith = datasets[5].read(
            1,
            window=window,
        )

        lulc = datasets[6].read(
            1,
            window=window,
        )


        valid_lith_codes = np.isin(
            np.rint(lith).astype(np.int16),
            np.arange(1, 12),
        )

        valid_lulc_codes = np.isin(
            np.rint(lulc).astype(np.int16),
            [10, 20, 30, 40, 50, 60, 80, 90],
        )


        valid &= valid_lith_codes
        valid &= valid_lulc_codes


        out = valid.astype(np.uint8)

        dst.write(
            out,
            1,
            window=window,
        )

        new_valid_count += int(
            valid.sum()
        )


# ---------------------------------------------------------------------
# Close predictors
# ---------------------------------------------------------------------

for ds in datasets:
    ds.close()


# ---------------------------------------------------------------------
# Compare against old 8-factor mask
# ---------------------------------------------------------------------

old_valid_count = 0

with rasterio.open(OLD_MASK) as ds:

    for _, window in ds.block_windows(1):

        old = ds.read(
            1,
            window=window,
        )

        old_valid_count += int(
            np.count_nonzero(old)
        )


difference = (
    new_valid_count
    - old_valid_count
)


print()
print("=" * 90)
print("VALID MASK RESULTS")
print("=" * 90)

print(
    "Old 8-factor mask valid cells:",
    f"{old_valid_count:,}",
)

print(
    "New 7-factor mask valid cells:",
    f"{new_valid_count:,}",
)

print(
    "Additional cells recovered:",
    f"{difference:,}",
)

print()
print("New mask:")
print(NEW_MASK)

print()
print("=" * 90)
print("MODEL A NO-ROAD VALID MASK COMPLETE ✅")
print("=" * 90)
```

```python
%%writefile scripts/91_modeling/predict_statewide_modelA_bias_controlled.py

from pathlib import Path
import json
import time

import numpy as np
import rasterio
import xgboost as xgb


# =====================================================================
# PATHS
# =====================================================================

ROOT = Path(
    "/content/Manipur_XGBoost_Colab"
)

RASTER_DIR = Path(
    "/content/Manipur_XGBoost_Statewide_Rasters"
)

MODEL_DIR = (
    ROOT
    / "output"
    / "road_matched_no_road_final_models"
)

OUT_DIR = (
    ROOT
    / "output"
    / "statewide_prediction_modelA_bias_controlled"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


MASK_PATH = (
    OUT_DIR
    / "manipur_modelA_no_road_common_valid_mask_30m.tif"
)


MEAN_OUT = (
    OUT_DIR
    / "manipur_modelA_bias_controlled_susceptibility_mean.tif"
)


SD_OUT = (
    OUT_DIR
    / "manipur_modelA_bias_controlled_prediction_sd.tif"
)


MANIFEST_OUT = (
    OUT_DIR
    / "modelA_bias_controlled_statewide_manifest.json"
)


# =====================================================================
# INPUT RASTERS
# =====================================================================

RASTER_PATHS = {

    "Elevation_m":
        RASTER_DIR
        / "manipur_dem_30m_utm46n.tif",

    "Slope_deg":
        RASTER_DIR
        / "manipur_slope_30m.tif",

    "TWI":
        RASTER_DIR
        / "manipur_twi_30m.tif",

    "Rainfall_CHELSA_Annual_mm":
        RASTER_DIR
        / "manipur_rainfall_annual_30m.tif",

    "Distance_to_River_m":
        RASTER_DIR
        / "manipur_distance_to_river_30m.tif",

    "Lithology":
        RASTER_DIR
        / "manipur_lithology_grouped_30m.tif",

    "LULC":
        RASTER_DIR
        / "manipur_lulc_worldcover_2021_30m.tif",
}


# =====================================================================
# MODEL FEATURE ORDER
# =====================================================================

FEATURES = [

    "Elevation_m",
    "Slope_deg",
    "TWI",
    "Rainfall_CHELSA_Annual_mm",
    "Distance_to_River_m",

    "Lithology_Group_1",
    "Lithology_Group_2",
    "Lithology_Group_3",
    "Lithology_Group_4",
    "Lithology_Group_5",
    "Lithology_Group_6",
    "Lithology_Group_7",
    "Lithology_Group_8",
    "Lithology_Group_9",
    "Lithology_Group_10",
    "Lithology_Group_11",

    "LULC_WorldCover_10",
    "LULC_WorldCover_20",
    "LULC_WorldCover_30",
    "LULC_WorldCover_40",
    "LULC_WorldCover_50",
    "LULC_WorldCover_60",
    "LULC_WorldCover_80",
    "LULC_WorldCover_90",
]


LITHOLOGY_CODES = list(
    range(1, 12)
)


LULC_CODES = [
    10,
    20,
    30,
    40,
    50,
    60,
    80,
    90,
]


# =====================================================================
# MODEL PATHS
# =====================================================================

MODEL_PATHS = [

    MODEL_DIR
    / f"manipur_xgb_road_matched_no_road_set_{i:02d}.json"

    for i in range(1, 6)
]


print("=" * 100)
print("STATEWIDE PREDICTION — BIAS-CONTROLLED MODEL A")
print("=" * 100)


# =====================================================================
# CHECK INPUTS
# =====================================================================

for name, path in RASTER_PATHS.items():

    if not path.exists():
        raise FileNotFoundError(
            f"Missing raster: {name}: {path}"
        )


if not MASK_PATH.exists():

    raise FileNotFoundError(
        f"Missing Model A mask: {MASK_PATH}"
    )


for model_path in MODEL_PATHS:

    if not model_path.exists():
        raise FileNotFoundError(
            f"Missing model: {model_path}"
        )


print("Input file check: PASS ✅")


# =====================================================================
# LOAD MODELS
# =====================================================================

models = []

for model_path in MODEL_PATHS:

    booster = xgb.Booster()

    booster.load_model(
        str(model_path)
    )

    models.append(
        booster
    )


print(
    f"Loaded models: {len(models)} ✅"
)


# =====================================================================
# OPEN RASTERS
# =====================================================================

datasets = {

    name:
        rasterio.open(path)

    for name, path
    in RASTER_PATHS.items()
}


mask_ds = rasterio.open(
    MASK_PATH
)


reference = datasets[
    "Elevation_m"
]


width = reference.width
height = reference.height
crs = reference.crs
transform = reference.transform


# =====================================================================
# ALIGNMENT CHECK
# =====================================================================

for name, ds in datasets.items():

    if ds.width != width or ds.height != height:

        raise RuntimeError(
            f"Shape mismatch: {name}"
        )


    if ds.crs != crs:

        raise RuntimeError(
            f"CRS mismatch: {name}"
        )


    if ds.transform != transform:

        raise RuntimeError(
            f"Transform mismatch: {name}"
        )


if (
    mask_ds.width != width
    or mask_ds.height != height
):

    raise RuntimeError(
        "Mask shape mismatch"
    )


if mask_ds.crs != crs:

    raise RuntimeError(
        "Mask CRS mismatch"
    )


if mask_ds.transform != transform:

    raise RuntimeError(
        "Mask transform mismatch"
    )


print("Raster alignment: PASS ✅")

print(
    "Shape:",
    (height, width),
)

print(
    "CRS:",
    crs,
)

print(
    "Resolution:",
    reference.res,
)


# =====================================================================
# OUTPUT PROFILE
# =====================================================================

NODATA = -9999.0


profile = reference.profile.copy()

profile.update(
    dtype="float32",
    count=1,
    nodata=NODATA,
    compress="deflate",
    predictor=3,
    tiled=True,
    blockxsize=512,
    blockysize=512,
    BIGTIFF="IF_SAFER",
)


# =====================================================================
# GLOBAL STATISTICS
# =====================================================================

predicted_count = 0

score_min = np.inf
score_max = -np.inf
score_sum = 0.0

sd_min = np.inf
sd_max = -np.inf
sd_sum = 0.0


start_time = time.time()


# =====================================================================
# STATEWIDE BLOCK-WISE PREDICTION
# =====================================================================

with rasterio.open(
    MEAN_OUT,
    "w",
    **profile,
) as mean_dst, rasterio.open(
    SD_OUT,
    "w",
    **profile,
) as sd_dst:


    windows = list(
        mask_ds.block_windows(1)
    )

    total_windows = len(
        windows
    )


    print()
    print(
        "Prediction windows:",
        total_windows,
    )


    for window_number, (_, window) in enumerate(
        windows,
        start=1,
    ):


        valid_mask = (
            mask_ds.read(
                1,
                window=window,
            )
            == 1
        )


        h = int(
            window.height
        )

        w = int(
            window.width
        )


        mean_block = np.full(
            (h, w),
            NODATA,
            dtype=np.float32,
        )


        sd_block = np.full(
            (h, w),
            NODATA,
            dtype=np.float32,
        )


        n_valid = int(
            valid_mask.sum()
        )


        if n_valid > 0:


            # =========================================================
            # READ CONTINUOUS FEATURES
            # =========================================================

            elevation = datasets[
                "Elevation_m"
            ].read(
                1,
                window=window,
            )[valid_mask]


            slope = datasets[
                "Slope_deg"
            ].read(
                1,
                window=window,
            )[valid_mask]


            twi = datasets[
                "TWI"
            ].read(
                1,
                window=window,
            )[valid_mask]


            rainfall = datasets[
                "Rainfall_CHELSA_Annual_mm"
            ].read(
                1,
                window=window,
            )[valid_mask]


            river = datasets[
                "Distance_to_River_m"
            ].read(
                1,
                window=window,
            )[valid_mask]


            # =========================================================
            # READ CATEGORICAL FEATURES
            # =========================================================

            lithology = np.rint(
                datasets[
                    "Lithology"
                ].read(
                    1,
                    window=window,
                )[valid_mask]
            ).astype(
                np.int16
            )


            lulc = np.rint(
                datasets[
                    "LULC"
                ].read(
                    1,
                    window=window,
                )[valid_mask]
            ).astype(
                np.int16
            )


            # =========================================================
            # CATEGORY VALIDATION
            # =========================================================

            if not np.all(
                np.isin(
                    lithology,
                    LITHOLOGY_CODES,
                )
            ):

                bad = np.unique(
                    lithology[
                        ~np.isin(
                            lithology,
                            LITHOLOGY_CODES,
                        )
                    ]
                )

                raise RuntimeError(
                    f"Unexpected lithology codes: {bad}"
                )


            if not np.all(
                np.isin(
                    lulc,
                    LULC_CODES,
                )
            ):

                bad = np.unique(
                    lulc[
                        ~np.isin(
                            lulc,
                            LULC_CODES,
                        )
                    ]
                )

                raise RuntimeError(
                    f"Unexpected LULC codes: {bad}"
                )


            # =========================================================
            # BUILD 24-FEATURE MATRIX
            # =========================================================

            X = np.zeros(
                (
                    n_valid,
                    len(FEATURES),
                ),
                dtype=np.float32,
            )


            # Continuous predictors

            X[:, 0] = elevation
            X[:, 1] = slope
            X[:, 2] = twi
            X[:, 3] = rainfall
            X[:, 4] = river


            # Lithology one-hot: columns 5–15

            for j, code in enumerate(
                LITHOLOGY_CODES
            ):

                X[
                    :,
                    5 + j
                ] = (
                    lithology == code
                ).astype(
                    np.float32
                )


            # LULC one-hot: columns 16–23

            for j, code in enumerate(
                LULC_CODES
            ):

                X[
                    :,
                    16 + j
                ] = (
                    lulc == code
                ).astype(
                    np.float32
                )


            # =========================================================
            # PREDICT WITH FIVE MODELS
            # =========================================================

            dmatrix = xgb.DMatrix(
                X,
                feature_names=FEATURES,
            )


            predictions = np.empty(
                (
                    n_valid,
                    len(models),
                ),
                dtype=np.float32,
            )


            for model_index, booster in enumerate(
                models
            ):

                predictions[
                    :,
                    model_index
                ] = booster.predict(
                    dmatrix
                )


            mean_prediction = predictions.mean(
                axis=1
            )


            sd_prediction = predictions.std(
                axis=1,
                ddof=0,
            )


            # =========================================================
            # SAFETY CHECK
            # =========================================================

            if not np.all(
                np.isfinite(
                    mean_prediction
                )
            ):

                raise RuntimeError(
                    "Non-finite mean predictions detected"
                )


            if not np.all(
                (
                    mean_prediction >= 0
                )
                & (
                    mean_prediction <= 1
                )
            ):

                raise RuntimeError(
                    "Prediction outside [0,1]"
                )


            # =========================================================
            # WRITE BLOCKS
            # =========================================================

            mean_block[
                valid_mask
            ] = mean_prediction


            sd_block[
                valid_mask
            ] = sd_prediction


            # =========================================================
            # UPDATE GLOBAL STATISTICS
            # =========================================================

            predicted_count += n_valid


            score_min = min(
                score_min,
                float(
                    mean_prediction.min()
                ),
            )


            score_max = max(
                score_max,
                float(
                    mean_prediction.max()
                ),
            )


            score_sum += float(
                mean_prediction.astype(
                    np.float64
                ).sum()
            )


            sd_min = min(
                sd_min,
                float(
                    sd_prediction.min()
                ),
            )


            sd_max = max(
                sd_max,
                float(
                    sd_prediction.max()
                ),
            )


            sd_sum += float(
                sd_prediction.astype(
                    np.float64
                ).sum()
            )


        mean_dst.write(
            mean_block,
            1,
            window=window,
        )


        sd_dst.write(
            sd_block,
            1,
            window=window,
        )


        if (
            window_number % 20 == 0
            or window_number == total_windows
        ):

            elapsed = (
                time.time()
                - start_time
            )

            print(
                f"Window "
                f"{window_number:>4}"
                f"/{total_windows}"
                f" | valid cells predicted: "
                f"{predicted_count:,}"
                f" | elapsed: "
                f"{elapsed:.1f}s"
            )


# =====================================================================
# CLOSE INPUTS
# =====================================================================

for ds in datasets.values():

    ds.close()


mask_ds.close()


# =====================================================================
# FINAL STATISTICS
# =====================================================================

elapsed_seconds = (
    time.time()
    - start_time
)


if predicted_count == 0:

    raise RuntimeError(
        "No valid cells predicted."
    )


score_mean = (
    score_sum
    / predicted_count
)


sd_mean = (
    sd_sum
    / predicted_count
)


# =====================================================================
# VALIDATE OUTPUT RASTERS
# =====================================================================

with rasterio.open(
    MEAN_OUT
) as mean_ds, rasterio.open(
    SD_OUT
) as uncertainty_ds:


    if (
        mean_ds.width != width
        or mean_ds.height != height
    ):

        raise RuntimeError(
            "Mean raster shape validation failed"
        )


    if (
        uncertainty_ds.width != width
        or uncertainty_ds.height != height
    ):

        raise RuntimeError(
            "SD raster shape validation failed"
        )


    if mean_ds.crs != crs:

        raise RuntimeError(
            "Mean raster CRS validation failed"
        )


    if uncertainty_ds.crs != crs:

        raise RuntimeError(
            "SD raster CRS validation failed"
        )


print()
print("=" * 100)
print("STATEWIDE MODEL A RESULTS")
print("=" * 100)

print(
    "Valid predicted cells:",
    f"{predicted_count:,}",
)

print()
print("Susceptibility mean raster:")

print(
    "  Minimum:",
    f"{score_min:.10f}",
)

print(
    "  Maximum:",
    f"{score_max:.10f}",
)

print(
    "  Mean:",
    f"{score_mean:.10f}",
)


print()
print("Ensemble SD raster:")

print(
    "  Minimum:",
    f"{sd_min:.10f}",
)

print(
    "  Maximum:",
    f"{sd_max:.10f}",
)

print(
    "  Mean:",
    f"{sd_mean:.10f}",
)


print()
print(
    "Elapsed seconds:",
    f"{elapsed_seconds:.2f}",
)


# =====================================================================
# MANIFEST
# =====================================================================

manifest = {

    "model":
        "Bias-Controlled Model A",

    "background_design":
        "Road-distance matched",

    "distance_to_road_predictor":
        False,

    "ensemble_models":
        5,

    "feature_count":
        len(FEATURES),

    "features":
        FEATURES,

    "valid_predicted_cells":
        predicted_count,

    "grid_shape":
        [
            height,
            width,
        ],

    "crs":
        str(crs),

    "resolution_m":
        30,

    "susceptibility_min":
        score_min,

    "susceptibility_max":
        score_max,

    "susceptibility_mean":
        score_mean,

    "ensemble_sd_min":
        sd_min,

    "ensemble_sd_max":
        sd_max,

    "ensemble_sd_mean":
        sd_mean,

    "elapsed_seconds":
        elapsed_seconds,

    "susceptibility_raster":
        str(MEAN_OUT),

    "uncertainty_raster":
        str(SD_OUT),

    "interpretation":
        (
            "Relative landslide susceptibility score. "
            "Not calibrated absolute landslide probability."
        ),
}


with open(
    MANIFEST_OUT,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        manifest,
        f,
        indent=2,
    )


print()
print("Created:")
print(MEAN_OUT)
print(SD_OUT)
print(MANIFEST_OUT)

print()
print("=" * 100)
print("STATEWIDE BIAS-CONTROLLED MODEL A PREDICTION COMPLETE ✅")
print("=" * 100)
```

```python
%%writefile scripts/91_modeling/validate_statewide_modelA_bias_controlled.py

from pathlib import Path

import numpy as np
import rasterio


ROOT = Path("/content/Manipur_XGBoost_Colab")

OUT_DIR = (
    ROOT
    / "output"
    / "statewide_prediction_modelA_bias_controlled"
)

MASK_PATH = (
    OUT_DIR
    / "manipur_modelA_no_road_common_valid_mask_30m.tif"
)

MEAN_PATH = (
    OUT_DIR
    / "manipur_modelA_bias_controlled_susceptibility_mean.tif"
)

SD_PATH = (
    OUT_DIR
    / "manipur_modelA_bias_controlled_prediction_sd.tif"
)


EXPECTED_VALID = 24_718_359


print("=" * 95)
print("VALIDATE STATEWIDE BIAS-CONTROLLED MODEL A")
print("=" * 95)


with rasterio.open(MASK_PATH) as mask_ds, \
     rasterio.open(MEAN_PATH) as mean_ds, \
     rasterio.open(SD_PATH) as sd_ds:

    # ---------------------------------------------------------------
    # Metadata checks
    # ---------------------------------------------------------------

    assert mean_ds.width == mask_ds.width == sd_ds.width
    assert mean_ds.height == mask_ds.height == sd_ds.height

    assert mean_ds.crs == mask_ds.crs == sd_ds.crs
    assert mean_ds.transform == mask_ds.transform == sd_ds.transform

    print("Shape:", (mean_ds.height, mean_ds.width))
    print("CRS:", mean_ds.crs)
    print("Resolution:", mean_ds.res)
    print("Transform:", mean_ds.transform)

    print()
    print("Metadata alignment: PASS ✅")


    # ---------------------------------------------------------------
    # Block-wise validation
    # ---------------------------------------------------------------

    mask_valid_count = 0
    mean_valid_count = 0
    sd_valid_count = 0

    invalid_inside_mean = 0
    invalid_inside_sd = 0

    prediction_outside_mask = 0

    score_min = np.inf
    score_max = -np.inf
    score_sum = 0.0

    sd_min = np.inf
    sd_max = -np.inf
    sd_sum = 0.0


    for _, window in mask_ds.block_windows(1):

        mask = (
            mask_ds.read(
                1,
                window=window,
            )
            == 1
        )

        mean = mean_ds.read(
            1,
            window=window,
        )

        sd = sd_ds.read(
            1,
            window=window,
        )


        mask_valid_count += int(
            mask.sum()
        )


        mean_valid = (
            mean != mean_ds.nodata
        )

        sd_valid = (
            sd != sd_ds.nodata
        )


        mean_valid_count += int(
            mean_valid.sum()
        )

        sd_valid_count += int(
            sd_valid.sum()
        )


        invalid_inside_mean += int(
            np.sum(
                mask
                & ~mean_valid
            )
        )

        invalid_inside_sd += int(
            np.sum(
                mask
                & ~sd_valid
            )
        )


        prediction_outside_mask += int(
            np.sum(
                (~mask)
                & (
                    mean_valid
                    | sd_valid
                )
            )
        )


        if np.any(mask):

            m = mean[mask]
            s = sd[mask]


            if not np.all(
                np.isfinite(m)
            ):
                raise RuntimeError(
                    "Non-finite susceptibility values detected."
                )


            if not np.all(
                np.isfinite(s)
            ):
                raise RuntimeError(
                    "Non-finite SD values detected."
                )


            if not np.all(
                (m >= 0)
                & (m <= 1)
            ):
                raise RuntimeError(
                    "Susceptibility values outside [0,1]."
                )


            if not np.all(
                s >= 0
            ):
                raise RuntimeError(
                    "Negative ensemble SD detected."
                )


            score_min = min(
                score_min,
                float(m.min()),
            )

            score_max = max(
                score_max,
                float(m.max()),
            )

            score_sum += float(
                m.astype(np.float64).sum()
            )


            sd_min = min(
                sd_min,
                float(s.min()),
            )

            sd_max = max(
                sd_max,
                float(s.max()),
            )

            sd_sum += float(
                s.astype(np.float64).sum()
            )


score_mean = (
    score_sum
    / mask_valid_count
)

sd_mean = (
    sd_sum
    / mask_valid_count
)


print()
print("=" * 95)
print("CELL VALIDATION")
print("=" * 95)

print(
    "Expected valid cells:",
    f"{EXPECTED_VALID:,}",
)

print(
    "Mask valid cells:",
    f"{mask_valid_count:,}",
)

print(
    "Susceptibility valid cells:",
    f"{mean_valid_count:,}",
)

print(
    "SD valid cells:",
    f"{sd_valid_count:,}",
)

print(
    "Missing susceptibility inside mask:",
    invalid_inside_mean,
)

print(
    "Missing SD inside mask:",
    invalid_inside_sd,
)

print(
    "Predictions outside valid mask:",
    prediction_outside_mask,
)


if mask_valid_count != EXPECTED_VALID:
    raise RuntimeError(
        "Unexpected mask valid-cell count."
    )


if mean_valid_count != EXPECTED_VALID:
    raise RuntimeError(
        "Unexpected susceptibility valid-cell count."
    )


if sd_valid_count != EXPECTED_VALID:
    raise RuntimeError(
        "Unexpected SD valid-cell count."
    )


if invalid_inside_mean != 0:
    raise RuntimeError(
        "Missing susceptibility values inside mask."
    )


if invalid_inside_sd != 0:
    raise RuntimeError(
        "Missing SD values inside mask."
    )


if prediction_outside_mask != 0:
    raise RuntimeError(
        "Predictions found outside valid mask."
    )


print()
print("Cell-mask consistency: PASS ✅")


print()
print("=" * 95)
print("RECALCULATED RASTER STATISTICS")
print("=" * 95)

print("Susceptibility:")
print(
    "  Minimum:",
    f"{score_min:.10f}",
)
print(
    "  Maximum:",
    f"{score_max:.10f}",
)
print(
    "  Mean:",
    f"{score_mean:.10f}",
)

print()

print("Ensemble SD:")
print(
    "  Minimum:",
    f"{sd_min:.10f}",
)
print(
    "  Maximum:",
    f"{sd_max:.10f}",
)
print(
    "  Mean:",
    f"{sd_mean:.10f}",
)


print()
print("=" * 95)
print("STATEWIDE MODEL A RASTER VALIDATION PASS ✅")
print("=" * 95)
```
